# Minimum Path Sum

**Problem Link:**  
https://leetcode.com/problems/minimum-path-sum/description/

## Problem Statement
Given an `m x n` grid filled with non-negative numbers, find a path from the top-left to the bottom-right corner that minimises the sum of all numbers along its path.

> You can only move **right** or **down** at any point.

## Examples

### Example 1
Input: `grid = [[1,3,1],[1,5,1],[4,2,1]]`  
Output: `7`  
Explanation: Path `1 → 3 → 1 → 1 → 1` minimises the sum.

### Example 2
Input: `grid = [[1,2,3],[4,5,6]]`  
Output: `12`

## Constraints
- `m == grid.length`
- `n == grid[i].length`
- `1 <= m, n <= 200`
- `0 <= grid[i][j] <= 200`


## Dynamic Programming — In-Place

### Strategy
Define `dp[i][j]` as the minimum path sum to reach cell `(i, j)` from `(0, 0)`.

- **Base case — top-left:** `dp[0][0] = grid[0][0]`
- **Base case — top row:** `dp[0][j] = dp[0][j-1] + grid[0][j]` (can only arrive from the left)
- **Base case — left column:** `dp[i][0] = dp[i-1][0] + grid[i][0]` (can only arrive from above)
- **Recurrence:** `dp[i][j] = grid[i][j] + min(dp[i-1][j], dp[i][j-1])`
- **Answer:** `dp[m-1][n-1]`

### Space Optimisation
We modify `grid` in-place instead of allocating a separate DP table — each cell is overwritten with its minimum path sum. This keeps space at **O(1)** extra (beyond the input).

### Time Complexity
- **O(m × n)** — every cell is visited exactly once

### Space Complexity
- **O(1)** extra — in-place modification of `grid`


```python
from typing import List

class Solution:
    def minPathSum(self, grid: List[List[int]]) -> int:
        m, n = len(grid), len(grid[0])

        # Fill top row (can only come from the left)
        for j in range(1, n):
            grid[0][j] += grid[0][j - 1]

        # Fill left column (can only come from above)
        for i in range(1, m):
            grid[i][0] += grid[i - 1][0]

        # Fill the rest: min of above or left, plus current cell
        for i in range(1, m):
            for j in range(1, n):
                grid[i][j] += min(grid[i - 1][j], grid[i][j - 1])

        return grid[m - 1][n - 1]
```

```python
import copy

def test_min_path_sum():
    sol = Solution()

    assert sol.minPathSum(copy.deepcopy([[1,3,1],[1,5,1],[4,2,1]])) == 7
    assert sol.minPathSum(copy.deepcopy([[1,2,3],[4,5,6]]))         == 12

    # Single cell
    assert sol.minPathSum([[5]])                                     == 5

    # Single row — only one path (go all right)
    assert sol.minPathSum([[1, 2, 3]])                               == 6

    # Single column — only one path (go all down)
    assert sol.minPathSum([[1],[2],[3]])                             == 6

    # All zeros
    assert sol.minPathSum([[0,0,0],[0,0,0]])                        == 0

    # Prefer going right then down vs down then right
    # [[1,100],[1,1]] → right path: 1+100+1=102, down path: 1+1+1=3
    assert sol.minPathSum([[1,100],[1,1]])                          == 3

    # 2x2 uniform grid
    assert sol.minPathSum([[1,2],[3,4]])                            == 7  # 1→2→4

    print("All test cases passed!")

test_min_path_sum()
```
