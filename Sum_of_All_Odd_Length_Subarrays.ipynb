# Sum of All Odd Length Subarrays

**Problem Link:**  
https://leetcode.com/problems/sum-of-all-odd-length-subarrays/

## Problem Statement
Given an array of positive integers `arr`, return the sum of all possible odd-length subarrays of `arr`.

A subarray is a contiguous subsequence of the array.

## Examples

### Example 1
Input: `arr = [1,4,2,5,3]`  
Output: `58`  
Explanation: The odd-length subarrays and their sums are:  
`[1]=1, [4]=4, [2]=2, [5]=5, [3]=3, [1,4,2]=7, [4,2,5]=11, [2,5,3]=10, [1,4,2,5,3]=15`  
Total: `1+4+2+5+3+7+11+10+15 = 58`

### Example 2
Input: `arr = [1,2]`  
Output: `3`  
Explanation: Only 2 subarrays of odd length exist — `[1]` and `[2]`. Their sum is `3`.

### Example 3
Input: `arr = [10,11,12]`  
Output: `66`

## Constraints
- `1 <= arr.length <= 100`
- `1 <= arr[i] <= 1000`

## Follow Up
Could you solve this problem in **O(n)** time complexity?


## Approach 1: Brute Force — Enumerate All Odd-Length Subarrays

### Strategy
- Use two nested loops: the outer loop fixes the starting index `i`, the inner loop steps through odd lengths `1, 3, 5, ...`
- For each `(start, length)` pair, sum the subarray `arr[i : i + length]` and add it to the total
- Stop when the subarray would exceed the array bounds

### Time Complexity
- **O(n³)** — O(n²) subarray pairs, each summed in O(n)

### Space Complexity
- **O(1)** extra space


```python
class SolutionBrute:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        total = 0
        n = len(arr)

        for i in range(n):
            for length in range(1, n - i + 1, 2):   # odd lengths only
                total += sum(arr[i : i + length])

        return total
```

## Approach 2: O(n) — Contribution of Each Element

### Strategy
Instead of enumerating subarrays, figure out how many odd-length subarrays each element `arr[i]` belongs to, then multiply.

For an element at index `i` in an array of length `n`:
- There are `left = i + 1` choices for the left endpoint (indices `0..i`)
- There are `right = n - i` choices for the right endpoint (indices `i..n-1`)
- Total subarrays containing `arr[i]` = `left * right`
- Among these, the ones with **odd length** = `(left * right + 1) // 2`
  - A subarray `[l..r]` has odd length iff `l` and `r` have the same parity
  - Counting same-parity pairs out of `left * right` total gives `ceil(left * right / 2)`

Sum `arr[i] * odd_count` over all `i` to get the answer in a single pass.

### Time Complexity
- **O(n)** — single pass through the array

### Space Complexity
- **O(1)** extra space


```python
class Solution:
    def sumOddLengthSubarrays(self, arr: list[int]) -> int:
        total = 0
        n = len(arr)

        for i in range(n):
            left  = i + 1        # number of choices for the left boundary
            right = n - i        # number of choices for the right boundary
            odd_count = (left * right + 1) // 2
            total += arr[i] * odd_count

        return total
```

```python
def test_sum_odd_length_subarrays():
    sol       = Solution()
    sol_brute = SolutionBrute()

    test_cases = [
        ([1, 4, 2, 5, 3], 58),
        ([1, 2],          3),
        ([10, 11, 12],    66),
        ([1],             1),
        ([1, 2, 3, 4, 5], 57),
    ]

    for arr, expected in test_cases:
        assert sol.sumOddLengthSubarrays(arr)       == expected, f"O(n) failed for {arr}"
        assert sol_brute.sumOddLengthSubarrays(arr) == expected, f"Brute failed for {arr}"

    print("All test cases passed!")

test_sum_odd_length_subarrays()
```
