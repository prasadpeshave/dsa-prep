# Minimum Cost to Move Chips to The Same Position

**Problem Link:**  
https://leetcode.com/problems/minimum-cost-to-move-chips-to-the-same-position/description/

## Problem Statement
We have `n` chips, where the position of the `i`th chip is `position[i]`.

We need to move all the chips to the same position. In one step, we can change the position of the `i`th chip from `position[i]` to:
- `position[i] + 2` or `position[i] - 2` with **cost = 0**
- `position[i] + 1` or `position[i] - 1` with **cost = 1**

Return the minimum cost needed to move all chips to the same position.

## Examples

### Example 1
Input: `position = [1,2,3]`  
Output: `1`  
Explanation: Move chip at position 3 → position 1 (cost 0). Move chip at position 2 → position 1 (cost 1). Total = 1.

### Example 2
Input: `position = [2,2,2,3,3]`  
Output: `2`  
Explanation: Move both chips at position 3 → position 2. Each costs 1. Total = 2.

### Example 3
Input: `position = [1,1000000000]`  
Output: `1`

## Constraints
- `1 <= position.length <= 100`
- `1 <= position[i] <= 10^9`


## Parity Counting

### Key Insight
- Moving a chip by **2** is **free** — so any chip can be moved to any position of the **same parity** at zero cost.
  - All chips at **even** positions can gather at any even position for free.
  - All chips at **odd** positions can gather at any odd position for free.
- The only cost arises when merging the two parity groups — moving one group onto the other costs **1 per chip**.
- Therefore the answer is simply: `min(count of odd-position chips, count of even-position chips)`
  - Move the smaller group onto the larger group, paying 1 per chip in the smaller group.

### Strategy
- Count how many chips are at odd positions (`odd`) and how many at even positions (`even`)
- Return `min(odd, even)`

### Time Complexity
- **O(n)** — single pass through `position`

### Space Complexity
- **O(1)** extra space


```python
from typing import List

class Solution:
    def minCostToMoveChips(self, position: List[int]) -> int:
        odd  = sum(p % 2 == 1 for p in position)
        even = sum(p % 2 == 0 for p in position)
        return min(odd, even)
```

```python
def test_min_cost_to_move_chips():
    sol = Solution()

    assert sol.minCostToMoveChips([1, 2, 3])          == 1
    assert sol.minCostToMoveChips([2, 2, 2, 3, 3])    == 2
    assert sol.minCostToMoveChips([1, 1000000000])    == 1

    # All chips at the same position — no cost needed
    assert sol.minCostToMoveChips([5, 5, 5])           == 0

    # All chips at odd positions — cost is 0 (free to merge)
    assert sol.minCostToMoveChips([1, 3, 5, 7])        == 0

    # All chips at even positions — cost is 0
    assert sol.minCostToMoveChips([2, 4, 6])           == 0

    # Balanced odd/even — move either group
    assert sol.minCostToMoveChips([1, 2, 3, 4])        == 2

    # Single chip — no moves needed
    assert sol.minCostToMoveChips([42])                == 0

    print("All test cases passed!")

test_min_cost_to_move_chips()
```
