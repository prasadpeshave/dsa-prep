# Largest Triangle Area

**Problem Link:**  
https://leetcode.com/problems/largest-triangle-area/

---

## Problem Statement

Given a list of points on a 2D plane, return the area of the largest triangle that can be formed using any three different points.

The area of a triangle formed by points A(x1,y1), B(x2,y2), C(x3,y3) can be computed using:

Area = | x1(y2 - y3) + x2(y3 - y1) + x3(y1 - y2) | / 2

## Example

Input: [[0,0],[0,1],[1,0],[0,2],[2,0]]

Output: 2.0

---

## Constraints

- 3 <= points.length <= 50
- Coordinates are between -50 and 50
- All points are unique

# Approach
---

## Brute Force (Try All Triplets)

### Strategy
- Iterate through all combinations of 3 distinct points
- Compute area using determinant formula
- Track maximum area

Since n <= 50, O(n³) is acceptable.

### Time Complexity
- **O(n³)**

### Space Complexity
- **O(1)**

```python
from typing import List

class Solution:
    def largestTriangleArea(self, points: List[List[int]]) -> float:
        n = len(points)
        max_area = 0.0

        for i in range(n):
            for j in range(i + 1, n):
                for k in range(j + 1, n):
                    x1, y1 = points[i]
                    x2, y2 = points[j]
                    x3, y3 = points[k]

                    area = abs(
                        x1*(y2 - y3) +
                        x2*(y3 - y1) +
                        x3*(y1 - y2)
                    ) / 2

                    max_area = max(max_area, area)

        return max_area
```

# Test Cases
---

```python
def test_largest_triangle_area():
    sol = Solution()

    assert abs(sol.largestTriangleArea([[0,0],[0,1],[1,0],[0,2],[2,0]]) - 2.0) < 1e-5
    assert abs(sol.largestTriangleArea([[1,0],[0,0],[0,1]]) - 0.5) < 1e-5

    print("All test cases passed!")

test_largest_triangle_area()
```
