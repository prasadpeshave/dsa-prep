## Problem Statement

Given an array of strings `words`, return all strings that are a substring of another word in the array.

You can return the answer in any order.

---

## Examples

### Example 1
Input: ["mass","as","hero","superhero"]  
Output: ["as","hero"]

### Example 2
Input: ["leetcode","et","code"]  
Output: ["et","code"]

### Example 3
Input: ["blue","green","bu"]  
Output: []

# Approach
---

## Brute Force (Substring Check)

### Strategy
- For each word, check against every other word
- If word is a substring of another word, add to result
- Avoid comparing word with itself

Since n ≤ 100 and length ≤ 30, O(n² * m) is acceptable.

### Time Complexity
- **O(n² * m)**

### Space Complexity
- **O(1)** (excluding output)

```python
from typing import List

class Solution:
    def stringMatching(self, words: List[str]) -> List[str]:
        result = []

        for i in range(len(words)):
            for j in range(len(words)):
                if i != j and words[i] in words[j]:
                    result.append(words[i])
                    break

        return result
```

# Test Cases
---

```python
def test_string_matching():
    sol = Solution()

    assert sorted(sol.stringMatching(["mass","as","hero","superhero"])) == ["as","hero"]
    assert sorted(sol.stringMatching(["leetcode","et","code"])) == ["code","et"]
    assert sol.stringMatching(["blue","green","bu"]) == []

    print('All test cases passed!')

test_string_matching()
```
