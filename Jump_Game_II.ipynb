# Jump Game II

**Problem Link:**  
https://leetcode.com/problems/jump-game-ii/description/

## Problem Statement
You are given a 0-indexed array of integers `nums` of length `n`. You are initially positioned at index `0`.

Each element `nums[i]` represents the maximum length of a forward jump from index `i`. In other words, if you are at index `i`, you can jump to any index `i + j` where `0 <= j <= nums[i]` and `i + j < n`.

Return the **minimum number of jumps** to reach index `n - 1`. It is guaranteed that you can always reach the last index.

## Examples

### Example 1
Input: `nums = [2,3,1,1,4]`  
Output: `2`  
Explanation: Jump from index 0 → 1 (1 step), then index 1 → 4 (3 steps). Total: 2 jumps.

### Example 2
Input: `nums = [2,3,0,1,4]`  
Output: `2`

## Constraints
- `1 <= nums.length <= 10⁴`
- `0 <= nums[i] <= 1000`
- It is guaranteed that you can reach `nums[n - 1]`


## Greedy — BFS-Level Expansion

### Strategy
Think of each jump count as a "level" in a BFS tree. At each level, we know the farthest index reachable with the current number of jumps. We greedily expand to the farthest reachable index within that level before incrementing the jump count.

- `jumps` — number of jumps taken so far
- `current_end` — the farthest index reachable with `jumps` jumps (the boundary of the current level)
- `farthest` — the farthest index reachable from any position in the current level

For each index `i` from `0` to `n-2`:
- Update `farthest = max(farthest, i + nums[i])`
- When `i == current_end`, we've exhausted the current level — increment `jumps` and set `current_end = farthest`
- We stop at `n-2` because reaching `n-1` doesn't require another jump from there

### Time Complexity
- **O(n)** — single pass through the array

### Space Complexity
- **O(1)** extra space


```python
from typing import List

class Solution:
    def jump(self, nums: List[int]) -> int:
        jumps       = 0
        current_end = 0
        farthest    = 0

        for i in range(len(nums) - 1):  # stop at n-2: no jump needed from last index
            farthest = max(farthest, i + nums[i])
            if i == current_end:        # reached the end of the current jump level
                jumps       += 1
                current_end  = farthest

        return jumps
```

```python
def test_jump():
    sol = Solution()

    assert sol.jump([2, 3, 1, 1, 4]) == 2
    assert sol.jump([2, 3, 0, 1, 4]) == 2

    # Single element — already at the last index, 0 jumps needed
    assert sol.jump([0])             == 0

    # Two elements — one jump always needed
    assert sol.jump([1, 0])          == 1

    # Can always jump to the end in one step
    assert sol.jump([5, 1, 1, 1, 1]) == 1

    # Must take every single step
    assert sol.jump([1, 1, 1, 1])    == 3

    # Large jump at start
    assert sol.jump([10, 1, 1, 1])   == 1

    # Classic greedy trap: greedy must look ahead, not just pick first available
    assert sol.jump([1, 2, 1, 1, 1]) == 3

    print("All test cases passed!")

test_jump()
```
