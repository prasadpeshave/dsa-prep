# Subsets II

**Problem Link:**  
https://leetcode.com/problems/subsets-ii/description/

## Problem Statement
Given an integer array `nums` that **may contain duplicates**, return all possible subsets (the power set).

The solution set must **not** contain duplicate subsets. Return the solution in any order.

## Examples

### Example 1
Input: `nums = [1,2,2]`  
Output: `[[],[1],[1,2],[1,2,2],[2],[2,2]]`

### Example 2
Input: `nums = [0]`  
Output: `[[],[0]]`

## Constraints
- `1 <= nums.length <= 10`
- `-10 <= nums[i] <= 10`


## Backtracking with Duplicate Pruning

### Difference from Subsets I
In Subsets I all elements are unique, so every branch produces a distinct subset. With duplicates, the same subset can be generated multiple times through different element orderings.

### Strategy
- **Sort** `nums` first — this groups duplicate values together, making it straightforward to detect and skip them
- Use the same `backtrack(start, current)` structure as Subsets I, recording at every node
- **Pruning rule:** within a single recursion level (same `start`), if `nums[i] == nums[i-1]` and `i > start`, skip `nums[i]` — it would generate the same subsets already produced by the previous identical element

### Why `i > start` (not `i > 0`)?
We only skip a duplicate if it appears at the **same level** of recursion. If the duplicate is the first element we're considering at this level (i.e. `i == start`), we must include it — otherwise we'd incorrectly skip subsets like `[2,2]` when the second `2` is the starting element.

### Time Complexity
- **O(2ⁿ × n)** in the worst case (all distinct) — with duplicates, significantly fewer branches are explored

### Space Complexity
- **O(n)** for the recursion stack and `current` list (excluding the output)


```python
from typing import List

class Solution:
    def subsetsWithDup(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        result = []

        def backtrack(start: int, current: list):
            result.append(list(current))   # record at every node
            for i in range(start, len(nums)):
                # Skip duplicate at the same recursion level
                if i > start and nums[i] == nums[i - 1]:
                    continue
                current.append(nums[i])
                backtrack(i + 1, current)
                current.pop()

        backtrack(0, [])
        return result
```

```python
def test_subsets_with_dup():
    sol = Solution()

    def normalize(subsets):
        return sorted([sorted(s) for s in subsets])

    # Example 1
    assert normalize(sol.subsetsWithDup([1,2,2])) == \
           normalize([[],[1],[1,2],[1,2,2],[2],[2,2]])

    # Example 2
    assert normalize(sol.subsetsWithDup([0])) == normalize([[],[0]])

    # No duplicates — same as Subsets I
    assert normalize(sol.subsetsWithDup([1,2,3])) == \
           normalize([[],[1],[2],[3],[1,2],[1,3],[2,3],[1,2,3]])

    # All duplicates
    assert normalize(sol.subsetsWithDup([2,2,2])) == normalize([[],[2],[2,2],[2,2,2]])

    # Two pairs of duplicates
    assert normalize(sol.subsetsWithDup([1,1,2,2])) == \
           normalize([[],[1],[1,1],[2],[1,2],[1,1,2],[2,2],[1,2,2],[1,1,2,2]])

    # No duplicate subsets in output
    result     = sol.subsetsWithDup([1,2,2])
    normalized = normalize(result)
    assert len(normalized) == len(set(map(tuple, normalized))), "Duplicate subsets found"

    # Empty set is always included
    assert [] in sol.subsetsWithDup([1,2,2])

    print("All test cases passed!")

test_subsets_with_dup()
```
