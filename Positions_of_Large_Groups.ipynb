# Positions of Large Groups

**Problem Link:**  
https://leetcode.com/problems/positions-of-large-groups/description/

## Problem Statement
In a string `s` of lowercase letters, these letters form consecutive groups of the same character.

For example, `s = "abbxxxxzyy"` has the groups `"a"`, `"bb"`, `"xxxx"`, `"z"`, and `"yy"`.

A group is identified by an interval `[start, end]`, where `start` and `end` denote the start and end indices (inclusive) of the group. In the above example, `"xxxx"` has the interval `[3, 6]`.

A group is considered **large** if it has **3 or more characters**.

Return the intervals of every large group sorted in increasing order by start index.

## Examples

### Example 1
Input: `s = "abbxxxxzzy"`  
Output: `[[3,6]]`  
Explanation: `"xxxx"` is the only large group with start index 3 and end index 6.

### Example 2
Input: `s = "abc"`  
Output: `[]`  
Explanation: Groups `"a"`, `"b"`, `"c"` — none are large.

### Example 3
Input: `s = "abcdddeeeeaabbbcd"`  
Output: `[[3,5],[6,9],[12,14]]`  
Explanation: The large groups are `"ddd"`, `"eeee"`, and `"bbb"`.

## Constraints
- `1 <= s.length <= 1000`
- `s` contains lowercase English letters only


## Two-Pointer / Sliding Window

### Strategy
- Use a pointer `start` to track the beginning of the current group
- Walk `i` from index `1` to `n` (one past the end to flush the last group)
- Whenever `s[i] != s[start]` (or `i` reaches the end of the string), the current group `[start, i-1]` is complete
  - If its length `i - start >= 3`, record `[start, i - 1]` as a large group
  - Advance `start` to `i` to begin tracking the next group
- The output is already sorted by start index since we scan left to right

### Time Complexity
- **O(n)** — single pass through the string

### Space Complexity
- **O(1)** extra space (excluding the output list)


```python
from typing import List

class Solution:
    def largeGroupPositions(self, s: str) -> List[List[int]]:
        result = []
        start  = 0
        n      = len(s)

        for i in range(1, n + 1):          # iterate one past the end to flush last group
            if i == n or s[i] != s[start]: # group boundary reached
                if i - start >= 3:         # large group check
                    result.append([start, i - 1])
                start = i                  # start tracking the next group

        return result
```

```python
def test_large_group_positions():
    sol = Solution()

    assert sol.largeGroupPositions("abbxxxxzzy")       == [[3, 6]]
    assert sol.largeGroupPositions("abc")              == []
    assert sol.largeGroupPositions("abcdddeeeeaabbbcd") == [[3, 5], [6, 9], [12, 14]]

    # Entire string is one large group
    assert sol.largeGroupPositions("aaaa")             == [[0, 3]]

    # No group reaches size 3
    assert sol.largeGroupPositions("aabbcc")           == []

    # Large group at the very end
    assert sol.largeGroupPositions("abzzz")            == [[2, 4]]

    # Single character string
    assert sol.largeGroupPositions("a")                == []

    # Multiple large groups back to back
    assert sol.largeGroupPositions("aaabbb")           == [[0, 2], [3, 5]]

    print("All test cases passed!")

test_large_group_positions()
```
