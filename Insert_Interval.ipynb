# Insert Interval

**Problem Link:**  
https://leetcode.com/problems/insert-interval/description/

## Problem Statement
You are given an array of non-overlapping intervals `intervals` sorted in ascending order by start value. You are also given a `newInterval = [start, end]`.

Insert `newInterval` into `intervals` such that the result is still sorted and has no overlapping intervals (merge if necessary).

## Examples

### Example 1
Input: `intervals = [[1,3],[6,9]], newInterval = [2,5]`  
Output: `[[1,5],[6,9]]`

### Example 2
Input: `intervals = [[1,2],[3,5],[6,7],[8,10],[12,16]], newInterval = [4,8]`  
Output: `[[1,2],[3,10],[12,16]]`  
Explanation: `[4,8]` overlaps with `[3,5]`, `[6,7]`, and `[8,10]` — all merged into `[3,10]`.

## Constraints
- `0 <= intervals.length <= 10⁴`
- `intervals[i].length == 2`
- `0 <= start_i <= end_i <= 10⁵`
- `intervals` is sorted by `start_i` in ascending order
- `0 <= start <= end <= 10⁵`


## Three-Phase Linear Scan

### Strategy
Since `intervals` is already sorted, we can handle the insertion in one pass with three distinct phases:

**Phase 1 — Add all intervals that end before `newInterval` starts:**  
While `intervals[i][1] < newInterval[0]` → no overlap, safely add to result as-is.

**Phase 2 — Merge all overlapping intervals into `newInterval`:**  
While `intervals[i][0] <= newInterval[1]` → overlap exists: expand `newInterval` to cover both.
- `newInterval[0] = min(newInterval[0], intervals[i][0])`
- `newInterval[1] = max(newInterval[1], intervals[i][1])`
After the loop, append the fully merged `newInterval`.

**Phase 3 — Add all remaining intervals (they start after `newInterval` ends):**  
Append the rest directly.

### Time Complexity
- **O(n)** — each interval is visited exactly once

### Space Complexity
- **O(n)** for the output list


```python
from typing import List

class Solution:
    def insert(self, intervals: List[List[int]], newInterval: List[int]) -> List[List[int]]:
        result = []
        i, n   = 0, len(intervals)

        # Phase 1: add all intervals that come entirely before newInterval
        while i < n and intervals[i][1] < newInterval[0]:
            result.append(intervals[i])
            i += 1

        # Phase 2: merge all overlapping intervals into newInterval
        while i < n and intervals[i][0] <= newInterval[1]:
            newInterval[0] = min(newInterval[0], intervals[i][0])
            newInterval[1] = max(newInterval[1], intervals[i][1])
            i += 1
        result.append(newInterval)

        # Phase 3: add all remaining intervals that come entirely after newInterval
        while i < n:
            result.append(intervals[i])
            i += 1

        return result
```

```python
def test_insert():
    sol = Solution()

    assert sol.insert([[1,3],[6,9]], [2,5])                             == [[1,5],[6,9]]
    assert sol.insert([[1,2],[3,5],[6,7],[8,10],[12,16]], [4,8])        == [[1,2],[3,10],[12,16]]

    # Empty intervals list
    assert sol.insert([], [2,5])                                        == [[2,5]]

    # newInterval before all existing
    assert sol.insert([[3,5],[6,9]], [1,2])                             == [[1,2],[3,5],[6,9]]

    # newInterval after all existing
    assert sol.insert([[1,3],[4,6]], [8,10])                            == [[1,3],[4,6],[8,10]]

    # newInterval covers all existing intervals
    assert sol.insert([[1,2],[3,4],[5,6]], [0,10])                      == [[0,10]]

    # newInterval touches boundary (no overlap)
    assert sol.insert([[1,3],[6,9]], [3,6])                             == [[1,9]]

    # Single interval that overlaps
    assert sol.insert([[1,5]], [2,3])                                   == [[1,5]]

    # No overlap — newInterval inserted between two intervals
    assert sol.insert([[1,2],[8,10]], [4,6])                            == [[1,2],[4,6],[8,10]]

    print("All test cases passed!")

test_insert()
```
