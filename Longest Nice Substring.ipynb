# Longest Nice Substring

**Problem Link:**  
https://leetcode.com/problems/longest-nice-substring/

---

## Problem Statement

A string is **nice** if for every letter it contains, both its uppercase and lowercase forms appear in the string.

Return the longest nice substring. If multiple exist, return the earliest one.
If none exist, return an empty string.

---

## Examples

### Example 1
Input: "YazaAay"  
Output: "aAa"

### Example 2
Input: "Bb"  
Output: "Bb"

### Example 3
Input: "c"  
Output: ""

# Approach
---

## Brute Force Substring Check

### Strategy
- Generate all substrings
- For each substring, check if it is nice
- Keep track of the longest valid substring

A substring is nice if for every character:
- `char.lower()` and `char.upper()` both exist in the substring

Since n ≤ 100, O(n³) is acceptable.

### Time Complexity
- **O(n³)**

### Space Complexity
- **O(n)**

```python
class Solution:
    def longestNiceSubstring(self, s: str) -> str:
        n = len(s)
        longest = ""

        for i in range(n):
            for j in range(i + 1, n + 1):
                substring = s[i:j]
                char_set = set(substring)

                if all(c.swapcase() in char_set for c in char_set):
                    if len(substring) > len(longest):
                        longest = substring

        return longest
```

# Test Cases
---

```python
def test_longest_nice_substring():
    sol = Solution()

    assert sol.longestNiceSubstring("YazaAay") == "aAa"
    assert sol.longestNiceSubstring("Bb") == "Bb"
    assert sol.longestNiceSubstring("c") == ""

    print("All test cases passed!")

test_longest_nice_substring()
```
