## Problem Statement
Write a solution to concatenate two DataFrames vertically into one DataFrame.

### DataFrame df1
| Column Name | Type   |
|-------------|--------|
| student_id  | int    |
| name        | object |
| age         | int    |

### DataFrame df2
| Column Name | Type   |
|-------------|--------|
| student_id  | int    |
| name        | object |
| age         | int    |

## Example 1

**Input:**

df1:
| student_id | name    | age |
|------------|---------|-----|
| 1          | Mason   | 8   |
| 2          | Ava     | 6   |
| 3          | Taylor  | 15  |
| 4          | Georgia | 17  |

df2:
| student_id | name | age |
|------------|------|-----|
| 5          | Leo  | 7   |
| 6          | Alex | 7   |

**Output:**
| student_id | name    | age |
|------------|---------|-----|
| 1          | Mason   | 8   |
| 2          | Ava     | 6   |
| 3          | Taylor  | 15  |
| 4          | Georgia | 17  |
| 5          | Leo     | 7   |
| 6          | Alex    | 7   |

**Explanation:** The two DataFrames are stacked vertically and their rows are combined.


## pd.concat — Vertical Stack

### Strategy
- Use `pd.concat()` with a list of the two DataFrames to stack them vertically (row-wise)
- Pass `ignore_index=True` to reset the index in the resulting DataFrame so there are no duplicate index values carried over from the originals
- Both DataFrames share the same columns, so no alignment issues arise

### Key Parameter
- `axis=0` (default) — concatenates along rows (vertical stack)
- `ignore_index=True` — resets the index to `0, 1, 2, ...` in the output

### Time Complexity
- **O(n + m)** where `n` and `m` are the number of rows in `df1` and `df2`

### Space Complexity
- **O(n + m)** for the resulting combined DataFrame


```python
import pandas as pd

def concatenateTables(df1: pd.DataFrame, df2: pd.DataFrame) -> pd.DataFrame:
    return pd.concat([df1, df2], ignore_index=True)
```

```python
def test_concatenate_tables():
    # Example 1 — from the problem
    df1 = pd.DataFrame({
        "student_id": [1, 2, 3, 4],
        "name":       ["Mason", "Ava", "Taylor", "Georgia"],
        "age":        [8, 6, 15, 17]
    })
    df2 = pd.DataFrame({
        "student_id": [5, 6],
        "name":       ["Leo", "Alex"],
        "age":        [7, 7]
    })

    result = concatenateTables(df1, df2)

    # Shape check
    assert result.shape == (6, 3), f"Expected shape (6, 3), got {result.shape}"

    # All student_ids present
    assert list(result["student_id"]) == [1, 2, 3, 4, 5, 6]

    # All names in correct order
    assert list(result["name"]) == ["Mason", "Ava", "Taylor", "Georgia", "Leo", "Alex"]

    # Index is reset (no duplicate indices)
    assert list(result.index) == list(range(6)), "Index should be reset to 0..5"

    # Columns unchanged
    assert list(result.columns) == ["student_id", "name", "age"]

    # Edge: concatenating with an empty DataFrame
    empty = pd.DataFrame(columns=["student_id", "name", "age"])
    result_empty = concatenateTables(df1, empty)
    assert result_empty.shape[0] == 4, "Concat with empty df should return df1 rows only"

    print("All test cases passed!")
    print()
    print("Result preview:")
    print(result.to_string(index=False))

test_concatenate_tables()
```
