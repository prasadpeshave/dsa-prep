# Pow(x, n)

**Problem Link:**  
https://leetcode.com/problems/powx-n/

## Problem Statement
Implement `pow(x, n)`, which calculates `x` raised to the power `n` (i.e., `xⁿ`).

## Examples

### Example 1
Input: `x = 2.00000, n = 10`  
Output: `1024.00000`

### Example 2
Input: `x = 2.10000, n = 3`  
Output: `9.26100`

### Example 3
Input: `x = 2.00000, n = -2`  
Output: `0.25000`  
Explanation: `2⁻² = 1/2² = 1/4 = 0.25`

## Constraints
- `-100.0 < x < 100.0`
- `-2³¹ <= n <= 2³¹ - 1`
- `n` is an integer
- Either `x` is not zero or `n > 0`
- `-10⁴ <= xⁿ <= 10⁴`


## Fast Power — Exponentiation by Squaring

### Strategy
A naive approach multiplies `x` by itself `n` times in O(n). With **exponentiation by squaring**, we halve the problem at every step:

- If `n` is **even**: `xⁿ = (x²)^(n/2)` — square `x` and halve `n`
- If `n` is **odd**: `xⁿ = x × x^(n-1)` — multiply by `x` once, then reduce to even case

This gives O(log n) multiplications.

### Handling Negative Exponents
- If `n < 0`: compute `x^|n|` then take the reciprocal: `result = 1 / result`

### Iterative Implementation
The iterative version avoids recursion stack overhead and is simpler to reason about:
- Work with `n = abs(n)`; track `result = 1.0`
- While `n > 0`:
  - If `n` is odd → multiply `result` by the current `x`
  - Square `x` and halve `n` each step

### Time Complexity
- **O(log n)** — `n` is halved at each step

### Space Complexity
- **O(1)** — iterative, no recursion stack


```python
class Solution:
    def myPow(self, x: float, n: int) -> float:
        negative = n < 0
        n        = abs(n)
        result   = 1.0

        while n > 0:
            if n % 2 == 1:       # odd exponent: multiply in the current base
                result *= x
            x  *= x              # square the base
            n  //= 2             # halve the exponent

        return 1 / result if negative else result
```

```python
def test_my_pow():
    sol = Solution()

    def approx(a, b, tol=1e-4):
        return abs(a - b) < tol

    assert approx(sol.myPow(2.0,    10),  1024.0)
    assert approx(sol.myPow(2.1,     3),  9.261)
    assert approx(sol.myPow(2.0,    -2),  0.25)

    # Zero exponent — always 1
    assert approx(sol.myPow(5.0,     0),  1.0)
    assert approx(sol.myPow(-3.0,    0),  1.0)

    # Exponent of 1
    assert approx(sol.myPow(7.5,     1),  7.5)

    # Negative base
    assert approx(sol.myPow(-2.0,    3), -8.0)
    assert approx(sol.myPow(-2.0,    4),  16.0)

    # Negative exponent
    assert approx(sol.myPow(4.0,    -1),  0.25)
    assert approx(sol.myPow(2.0,   -10),  1/1024)

    # x = 1 (always 1)
    assert approx(sol.myPow(1.0,  10000),  1.0)

    # Large n (stress test for O(log n))
    assert approx(sol.myPow(1.0, 2**30),  1.0)

    print("All test cases passed!")

test_my_pow()
```
