# Rotate String

**Problem Link:**

---

## Problem Statement

Given two strings `s` and `goal`, return `True` if and only if `s` can become `goal` after some number of left shifts.

A shift consists of moving the leftmost character to the rightmost position.

## Examples

### Example 1
Input: `s = "abcde"`, `goal = "cdeab"`  
Output: `True`

### Example 2
Input: `s = "abcde"`, `goal = "abced"`  
Output: `False`

---

## Constraints

- `1 <= s.length, goal.length <= 100`
- Only lowercase English letters

# Approach
---

## String Trick

### Key Insight
If `goal` is a rotation of `s`, then it must be a substring of `s + s`.

Example:
`s = abcde`

`s + s = abcdeabcde`

All possible rotations of `s` are substrings of `s + s`.

### Steps
1. Check if lengths are equal.
2. Check if `goal` is in `s + s`.

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)**

```python
class Solution:
    def rotateString(self, s: str, goal: str) -> bool:
        if len(s) != len(goal):
            return False
        return goal in (s + s)
```

# Test Cases
---

```python
def test_rotate_string():
    sol = Solution()

    assert sol.rotateString("abcde", "cdeab") == True
    assert sol.rotateString("abcde", "abced") == False
    assert sol.rotateString("a", "a") == True
    assert sol.rotateString("aa", "aa") == True

    print("All test cases passed!")

test_rotate_string()
```
