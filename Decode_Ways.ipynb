# 91. Decode Ways

Problem Link: https://leetcode.com/problems/decode-ways/

## Problem Summary
Given a string containing only digits, return the number of ways to decode it.

Mapping:
1 -> A, 2 -> B, ..., 26 -> Z

Rules:
- '0' cannot be decoded alone
- Valid two-digit numbers are between 10 and 26


## Approach (Dynamic Programming)

Let dp[i] represent the number of ways to decode substring s[0:i]

Transitions:
- If s[i] != '0', add dp[i-1]
- If 10 <= int(s[i-1:i+1]) <= 26, add dp[i-2]

We optimize space by keeping only two variables.

```python
class Solution:
    def numDecodings(self, s: str) -> int:
        if not s or s[0] == '0':
            return 0
        
        prev2 = 1
        prev1 = 1
        
        for i in range(1, len(s)):
            curr = 0
            
            if s[i] != '0':
                curr += prev1
            
            two_digit = int(s[i-1:i+1])
            if 10 <= two_digit <= 26:
                curr += prev2
            
            prev2 = prev1
            prev1 = curr
        
        return prev1
```

## Complexity
- Time: O(n)
- Space: O(1)


```python
sol = Solution()
print(sol.numDecodings("12"))
print(sol.numDecodings("226"))
print(sol.numDecodings("06"))
print(sol.numDecodings("11106"))
```
