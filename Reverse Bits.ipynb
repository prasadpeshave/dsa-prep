# Reverse Bits

**Problem Link:**  
https://leetcode.com/problems/reverse-bits/

## Problem Statement
Reverse the bits of a given **32-bit signed integer**.

## Examples

### Example 1
Input: `n = 43261596`  
Output: `964176192`

Binary:
43261596 -> 00000010100101000001111010011100
964176192 -> 00111001011110000010100101000000


### Example 2
Input: `n = 2147483644`  
Output: `1073741822`

## Constraints
- `0 <= n <= 2^31 - 2`
- `n` is even



## Bit-by-Bit Reversal

### Strategy
- Initialize result = 0
- Iterate exactly 32 times
- In each step:
  - Shift result left by 1
  - Add the least significant bit of `n`
  - Shift `n` right by 1

### Why This Works
- We extract bits from right to left
- Build the reversed number from left to right
- Fixed 32 iterations ensure correctness

### Time Complexity
- **O(32)** → constant time

### Space Complexity
- **O(1)**


```python
class Solution:
    def reverseBits(self, n: int) -> int:
        result = 0
        for _ in range(32):
            result = (result << 1) | (n & 1)
            n >>= 1
        return result
```

```python
def test_reverse_bits():
    sol = Solution()
    assert sol.reverseBits(43261596) == 964176192
    assert sol.reverseBits(2147483644) == 1073741822
    assert sol.reverseBits(0) == 0
    assert sol.reverseBits(1) == 2147483648
    print("All test cases passed!")

test_reverse_bits()
```
