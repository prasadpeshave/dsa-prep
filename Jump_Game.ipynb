# Jump Game

**Problem Link:**  
https://leetcode.com/problems/jump-game/description/

## Problem Statement
You are given an integer array `nums`. You are initially positioned at the array's first index, and each element represents your maximum jump length at that position.

Return `true` if you can reach the last index, or `false` otherwise.

## Examples

### Example 1
Input: `nums = [2,3,1,1,4]`  
Output: `true`  
Explanation: Jump 1 step from index 0 → 1, then 3 steps to the last index.

### Example 2
Input: `nums = [3,2,1,0,4]`  
Output: `false`  
Explanation: Every path leads to index 3 where `nums[3] = 0`, making it impossible to proceed.

## Constraints
- `1 <= nums.length <= 10⁴`
- `0 <= nums[i] <= 10⁵`


## Greedy — Track Maximum Reachable Index

### Strategy
- Maintain `max_reach`: the farthest index reachable so far
- Iterate through each index `i`:
  - If `i > max_reach`, the current index is unreachable — return `False` immediately
  - Otherwise, update `max_reach = max(max_reach, i + nums[i])`
  - If `max_reach >= n - 1` at any point, the last index is reachable — return `True`
- If the loop completes without returning `False`, return `True`

### Why Greedy Works
We never need to track which specific path we take — only the farthest point reachable. If at any index we can't even stand on it (i.e. it's beyond `max_reach`), then no path exists to proceed.

### Time Complexity
- **O(n)** — single pass through the array

### Space Complexity
- **O(1)** extra space


```python
from typing import List

class Solution:
    def canJump(self, nums: List[int]) -> bool:
        max_reach = 0

        for i in range(len(nums)):
            if i > max_reach:
                return False          # current index is unreachable
            max_reach = max(max_reach, i + nums[i])
            if max_reach >= len(nums) - 1:
                return True

        return True
```

```python
def test_can_jump():
    sol = Solution()

    assert sol.canJump([2, 3, 1, 1, 4]) == True
    assert sol.canJump([3, 2, 1, 0, 4]) == False

    # Single element — already at the last index
    assert sol.canJump([0])             == True

    # Zero at start with more elements
    assert sol.canJump([0, 1])          == False

    # Can just barely reach the end
    assert sol.canJump([1, 0])          == True
    assert sol.canJump([2, 0, 0])       == True

    # Blocked by a zero in the middle
    assert sol.canJump([1, 0, 1])       == False

    # All zeros except the first
    assert sol.canJump([5, 0, 0, 0, 0]) == True

    # Long chain of ones
    assert sol.canJump([1] * 10)        == True

    # Zero before the last element
    assert sol.canJump([3, 0, 0, 1, 0]) == True   # jump straight to index 3 then 4

    print("All test cases passed!")

test_can_jump()
```
