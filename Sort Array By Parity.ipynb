# Sort Array By Parity

## Problem Statement

Given an integer array `nums`, move all even integers to the beginning followed by all odd integers.

Return any valid array satisfying this condition.

---

## Examples

### Example 1
Input: nums = [3,1,2,4]  
Output: [2,4,3,1]

### Example 2
Input: nums = [0]  
Output: [0]

# Approach
---

## Two Pointer (In-Place Partition)

### Strategy
- Maintain pointer `left` for placing even numbers
- Traverse array
- When even number found → swap with `left`
- Increment `left`

### Time Complexity
- **O(n)**

### Space Complexity
- **O(1)**

```python
from typing import List

class Solution:
    def sortArrayByParity(self, nums: List[int]) -> List[int]:
        left = 0

        for i in range(len(nums)):
            if nums[i] % 2 == 0:
                nums[left], nums[i] = nums[i], nums[left]
                left += 1

        return nums
```

# Test Cases
---

```python
def test_sort_array_by_parity():
    sol = Solution()

    result1 = sol.sortArrayByParity([3,1,2,4])
    assert all(x % 2 == 0 for x in result1[:2])
    assert sol.sortArrayByParity([0]) == [0]

    print('All test cases passed!')

test_sort_array_by_parity()
```
