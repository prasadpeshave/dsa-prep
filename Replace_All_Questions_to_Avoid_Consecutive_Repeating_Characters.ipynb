# Replace All ?'s to Avoid Consecutive Repeating Characters

**Problem Link:**  
https://leetcode.com/problems/replace-all-s-to-avoid-consecutive-repeating-characters/description/

## Problem Statement
Given a string `s` containing only lowercase English letters and the `'?'` character, convert all the `'?'` characters into lowercase letters such that the final string does not contain any consecutive repeating characters. You cannot modify the non-`'?'` characters.

It is guaranteed that there are no consecutive repeating characters in the given string except for `'?'`.

Return the final string after all the conversions (possibly zero) have been made. If there is more than one solution, return any of them. It can be shown that an answer is always possible with the given constraints.

## Examples

### Example 1
Input: `s = "?zs"`  
Output: `"azs"`  
Explanation: There are 25 solutions for this problem. From `"azs"` to `"yzs"`, all are valid. Only `"z"` is an invalid modification as the string would consist of consecutive repeating characters in `"zzs"`.

### Example 2
Input: `s = "ubv?w"`  
Output: `"ubvaw"`  
Explanation: There are 24 solutions for this problem. Only `"v"` and `"w"` are invalid modifications as the strings would consist of consecutive repeating characters in `"ubvvw"` and `"ubvww"`.

## Constraints
- `1 <= s.length <= 100`
- `s` consists of lowercase English letters and `'?'`


## Greedy Character Replacement

### Strategy
- Convert the string to a list for mutability
- Iterate through each character; skip non-`'?'` characters
- For each `'?'`, try letters `'a'`, `'b'`, `'c'` in order
- Assign the first letter that differs from both the left neighbor (`i - 1`) and the right neighbor (`i + 1`)
- Only three candidates are ever needed — with at most two neighbors, at least one of `'a'`, `'b'`, `'c'` is always safe
- Join and return the list as a string

### Time Complexity
- **O(n)** where `n` is the length of the string — each character is visited once, and the inner loop runs at most 3 times

### Space Complexity
- **O(n)** for converting the string to a mutable list


```python
class Solution:
    def modifyString(self, s: str) -> str:
        s = list(s)

        for i in range(len(s)):
            if s[i] == '?':
                for candidate in 'abc':
                    left  = s[i - 1] if i > 0 else None
                    right = s[i + 1] if i < len(s) - 1 else None
                    if candidate != left and candidate != right:
                        s[i] = candidate
                        break

        return "".join(s)
```

```python
def no_consecutive_repeats(s: str) -> bool:
    return all(s[i] != s[i + 1] for i in range(len(s) - 1))

def test_modify_string():
    sol = Solution()

    result = sol.modifyString("?zs")
    assert '?' not in result and no_consecutive_repeats(result), f"Failed for '?zs': got {result}"

    result = sol.modifyString("ubv?w")
    assert '?' not in result and no_consecutive_repeats(result), f"Failed for 'ubv?w': got {result}"

    result = sol.modifyString("j?qg??b")
    assert '?' not in result and no_consecutive_repeats(result), f"Failed for 'j?qg??b': got {result}"

    result = sol.modifyString("??")
    assert '?' not in result and no_consecutive_repeats(result), f"Failed for '??': got {result}"

    result = sol.modifyString("abc")
    assert result == "abc", f"Failed for 'abc' (no ?s): got {result}"

    result = sol.modifyString("?")
    assert '?' not in result and len(result) == 1, f"Failed for '?': got {result}"

    print("All test cases passed!")

test_modify_string()
```
