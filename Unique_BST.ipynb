# 96. Unique Binary Search Trees

Problem Link: https://leetcode.com/problems/unique-binary-search-trees/

## Problem Summary
Given n nodes with values from 1 to n, return the number of structurally unique BSTs.

This is a classic Catalan number problem.


## Approach (Dynamic Programming)

Let dp[i] represent the number of unique BSTs with i nodes.

For each root j (1 to i):
- Left subtree has j-1 nodes
- Right subtree has i-j nodes

dp[i] += dp[j-1] * dp[i-j]

Base case:
- dp[0] = 1
- dp[1] = 1


```python
class Solution:
    def numTrees(self, n: int) -> int:
        dp = [0] * (n + 1)
        dp[0] = 1
        dp[1] = 1

        for nodes in range(2, n + 1):
            total = 0
            for root in range(1, nodes + 1):
                left = dp[root - 1]
                right = dp[nodes - root]
                total += left * right
            dp[nodes] = total

        return dp[n]
```

## Complexity
- Time: O(n^2)
- Space: O(n)


```python
sol = Solution()
print(sol.numTrees(3))
print(sol.numTrees(1))
```
