## Problem Statement

Given a sorted array `nums` and a `target`, return its index if found.
Otherwise, return -1.

You must achieve **O(log n)** time complexity.

---

## Examples

Input: nums = [-1,0,3,5,9,12], target = 9  
Output: 4

Input: nums = [-1,0,3,5,9,12], target = 2  
Output: -1

# Approach
---

## Classic Binary Search

### Strategy
- Maintain two pointers: `left` and `right`
- Compute middle index
- Compare middle element with target
- Narrow search space accordingly

### Time Complexity
- **O(log n)**

### Space Complexity
- **O(1)**

```python
from typing import List

class Solution:
    def search(self, nums: List[int], target: int) -> int:
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = left + (right - left) // 2

            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return -1
```

# Test Cases
---

```python
def test_binary_search():
    sol = Solution()

    assert sol.search([-1,0,3,5,9,12], 9) == 4
    assert sol.search([-1,0,3,5,9,12], 2) == -1
    assert sol.search([1], 1) == 0
    assert sol.search([1], 0) == -1

    print('All test cases passed!')

test_binary_search()
```
