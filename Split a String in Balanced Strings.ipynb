# Split a String in Balanced Strings

## Problem Link
https://leetcode.com/problems/split-a-string-in-balanced-strings/

---

## Problem Statement
A balanced string has an equal number of **'L'** and **'R'** characters.

Given a balanced string `s`, split it into the **maximum number of balanced substrings**.

Return the maximum number of balanced substrings.

## Approach

### Greedy Counting
- Traverse the string
- Maintain a counter
  - `+1` for `'R'`
  - `-1` for `'L'`
- Whenever the counter becomes **0**, we found a balanced substring
- Increment the result counter

This greedy approach works because the earliest valid split always maximizes the number of substrings.

### Time Complexity
- **O(n)**

### Space Complexity
- **O(1)**

```python
class Solution:
    def balancedStringSplit(self, s: str) -> int:
        balance = 0
        count = 0

        for ch in s:
            if ch == 'R':
                balance += 1
            else:
                balance -= 1

            if balance == 0:
                count += 1

        return count
```

## Test Cases

```python
def test_balanced_string_split():
    sol = Solution()

    assert sol.balancedStringSplit('RLRRLLRLRL') == 4
    assert sol.balancedStringSplit('RLRRRLLRLL') == 2
    assert sol.balancedStringSplit('LLLLRRRR') == 1

    print('All test cases passed!')

test_balanced_string_split()
```
