# Isomorphic Strings

## Problem Statement

Given two strings `s` and `t`, determine if they are isomorphic.

Two strings are isomorphic if characters in `s` can be replaced to get `t` with a one-to-one mapping (bijection).

No two characters may map to the same character.

---

## Examples

### Example 1
Input: s = "egg", t = "add"  
Output: True

### Example 2
Input: s = "foo", t = "bar"  
Output: False

### Example 3
Input: s = "paper", t = "title"  
Output: True

# Approach
---

## HashMap (Two Dictionaries)

### Strategy
- Use two dictionaries:
  - `map_s_t` to map characters from s to t
  - `map_t_s` to ensure reverse mapping is also unique
- Traverse both strings together
- Validate mapping consistency

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)**

```python
class Solution:
    def isIsomorphic(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        map_s_t = {}
        map_t_s = {}

        for ch1, ch2 in zip(s, t):
            if ch1 in map_s_t and map_s_t[ch1] != ch2:
                return False
            if ch2 in map_t_s and map_t_s[ch2] != ch1:
                return False

            map_s_t[ch1] = ch2
            map_t_s[ch2] = ch1

        return True
```

# Test Cases
---

```python
def test_isomorphic():
    sol = Solution()

    assert sol.isIsomorphic("egg", "add") == True
    assert sol.isIsomorphic("foo", "bar") == False
    assert sol.isIsomorphic("paper", "title") == True
    assert sol.isIsomorphic("ab", "aa") == False

    print('All test cases passed!')

test_isomorphic()
```
