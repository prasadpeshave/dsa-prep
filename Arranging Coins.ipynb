# Arranging Coins

## Problem Link
https://leetcode.com/problems/arranging-coins/

---

## Problem Statement
You have `n` coins and want to build a staircase where the `i-th` row contains exactly `i` coins.

Return the number of **complete rows** that can be formed.

## Approach

### Binary Search
Coins required for `k` rows = `k * (k + 1) / 2`.

We find the largest `k` such that:
`k * (k + 1) / 2 <= n`.

Binary search helps find this efficiently.

### Time Complexity
- O(log n)

### Space Complexity
- O(1)

```python
class Solution:
    def arrangeCoins(self, n: int) -> int:
        left, right = 0, n

        while left <= right:
            mid = (left + right) // 2
            coins = mid * (mid + 1) // 2

            if coins == n:
                return mid
            elif coins < n:
                left = mid + 1
            else:
                right = mid - 1

        return right
```

## Test Cases

```python
def test_arrange_coins():
    sol = Solution()

    assert sol.arrangeCoins(5) == 2
    assert sol.arrangeCoins(8) == 3
    assert sol.arrangeCoins(1) == 1

    print('All tests passed!')

test_arrange_coins()
```
