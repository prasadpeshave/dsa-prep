# Letter Combinations of a Phone Number

**Problem Link:**  
https://leetcode.com/problems/letter-combinations-of-a-phone-number/description/

## Problem Statement
Given a string containing digits from `2-9` inclusive, return all possible letter combinations that the number could represent. Return the answer in any order.

A mapping of digits to letters (just like on telephone buttons) is given below. Note that `1` does not map to any letters.

```
2 → abc    3 → def
4 → ghi    5 → jkl    6 → mno
7 → pqrs   8 → tuv    9 → wxyz
```

## Examples

### Example 1
Input: `digits = "23"`  
Output: `["ad","ae","af","bd","be","bf","cd","ce","cf"]`

### Example 2
Input: `digits = ""`  
Output: `[]`

### Example 3
Input: `digits = "2"`  
Output: `["a","b","c"]`

## Constraints
- `0 <= digits.length <= 4`
- `digits[i]` is a digit in the range `['2', '9']`


## Backtracking

### Strategy
- Define the phone keypad mapping as a dictionary
- Use a recursive backtracking function `backtrack(index, current)` where:
  - `index` is the current position in `digits`
  - `current` is the combination built so far
- **Base case:** when `index == len(digits)`, `current` is a complete combination — add it to the result
- **Recursive case:** for each letter mapped to `digits[index]`, append it to `current` and recurse on `index + 1`
- Return an empty list immediately if `digits` is empty

### Why Backtracking?
- We explore all possible combinations by building them character by character
- At each step we branch once per letter for the current digit, then "undo" (backtrack) by popping the letter and trying the next one
- The call tree has depth `len(digits)` and branching factor at most 4 (for digit `7` or `9`)

### Time Complexity
- **O(4ⁿ × n)** where `n` is the number of digits — at most 4 choices per digit, and each combination takes O(n) to build

### Space Complexity
- **O(n)** for the recursion call stack (depth = number of digits)


````python
from typing import List

class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        if not digits:
            return []

        phone_map = {
            '2': 'abc', '3': 'def',
            '4': 'ghi', '5': 'jkl', '6': 'mno',
            '7': 'pqrs', '8': 'tuv', '9': 'wxyz'
        }

        result = []

        def backtrack(index: int, current: list):
            if index == len(digits):
                result.append("".join(current))
                return
            for letter in phone_map[digits[index]]:
                current.append(letter)
                backtrack(index + 1, current)
                current.pop()

        backtrack(0, [])
        return result
````

````python
def test_letter_combinations():
    sol = Solution()

    # Example 1
    assert sorted(sol.letterCombinations("23")) == sorted(["ad","ae","af","bd","be","bf","cd","ce","cf"])

    # Example 2: empty input
    assert sol.letterCombinations("") == []

    # Example 3: single digit
    assert sorted(sol.letterCombinations("2")) == ["a", "b", "c"]

    # Digit 9 has 4 letters
    assert sorted(sol.letterCombinations("9")) == ["w", "x", "y", "z"]

    # Two digits — check count: 3 × 4 = 12 combinations
    assert len(sol.letterCombinations("29")) == 12

    # Three digits — 3 × 3 × 3 = 27 combinations
    assert len(sol.letterCombinations("234")) == 27

    # Four digits with 4-letter keys — 4 × 4 = 16 for "79"
    assert len(sol.letterCombinations("79")) == 16

    # All combinations are unique
    result = sol.letterCombinations("23")
    assert len(result) == len(set(result)), "Duplicate combinations found"

    print("All test cases passed!")

test_letter_combinations()
````
