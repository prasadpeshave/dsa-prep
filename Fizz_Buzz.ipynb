# Fizz Buzz

**Problem Link:**  
https://leetcode.com/problems/fizz-buzz/

## Problem Statement
Given an integer `n`, return a string array `answer` (1-indexed) where:
- `answer[i] == "FizzBuzz"` if `i` is divisible by `3` and `5`
- `answer[i] == "Fizz"` if `i` is divisible by `3`
- `answer[i] == "Buzz"` if `i` is divisible by `5`
- `answer[i] == i` (as a string) if none of the above conditions are true

## Examples

### Example 1
Input: `n = 3`  
Output: `["1","2","Fizz"]`

### Example 2
Input: `n = 5`  
Output: `["1","2","Fizz","4","Buzz"]`

### Example 3
Input: `n = 15`  
Output: `["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]`

## Constraints
- `1 <= n <= 10^4`


## String Concatenation (Scalable Approach)

### Strategy
- Iterate through numbers `1` to `n`
- For each number, **build the result string by concatenation** rather than chaining `if/elif` blocks
  - Start with an empty string `token`
  - Append `"Fizz"` if divisible by `3`
  - Append `"Buzz"` if divisible by `5`
  - If `token` is still empty, use the number itself as a string
- This approach scales cleanly to any number of new divisor rules (e.g. `"Jazz"` for 7) without adding more `elif` branches

### Time Complexity
- **O(n)** — single pass through `1` to `n`

### Space Complexity
- **O(n)** for the output array


```python
from typing import List

class Solution:
    def fizzBuzz(self, n: int) -> List[str]:
        result = []

        for i in range(1, n + 1):
            token = ""
            if i % 3 == 0:
                token += "Fizz"
            if i % 5 == 0:
                token += "Buzz"
            result.append(token or str(i))

        return result
```

```python
def test_fizz_buzz():
    sol = Solution()

    assert sol.fizzBuzz(3)  == ["1", "2", "Fizz"]
    assert sol.fizzBuzz(5)  == ["1", "2", "Fizz", "4", "Buzz"]
    assert sol.fizzBuzz(15) == [
        "1", "2", "Fizz", "4", "Buzz",
        "Fizz", "7", "8", "Fizz", "Buzz",
        "11", "Fizz", "13", "14", "FizzBuzz"
    ]

    # Edge: n = 1
    assert sol.fizzBuzz(1) == ["1"]

    # Spot-check: 30 must be "FizzBuzz"
    assert sol.fizzBuzz(30)[29] == "FizzBuzz"

    # Spot-check: 7 must be "7"
    assert sol.fizzBuzz(10)[6] == "7"

    print("All test cases passed!")

test_fizz_buzz()
```
