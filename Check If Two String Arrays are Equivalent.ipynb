# Check If Two String Arrays are Equivalent

**Problem Link:**  
https://leetcode.com/problems/check-if-two-string-arrays-are-equivalent/

---

## Problem Statement

Given two string arrays `word1` and `word2`, return `True` if they represent the same string.

An array represents a string if all elements concatenated in order form that string.

## Examples

### Example 1
Input: `word1 = ["ab", "c"]`, `word2 = ["a", "bc"]`  
Output: `True`

### Example 2
Input: `word1 = ["a", "cb"]`, `word2 = ["ab", "c"]`  
Output: `False`

### Example 3
Input: `word1 = ["abc", "d", "defg"]`, `word2 = ["abcddefg"]`  
Output: `True`

---

## Constraints

- `1 <= word1.length, word2.length <= 10^3`
- Total characters <= `10^3`
- Only lowercase letters

# Approach
---

## Simple Concatenation

### Strategy
- Join all elements of `word1` into one string
- Join all elements of `word2` into one string
- Compare the resulting strings

### Time Complexity
- **O(n)** where `n` is total characters

### Space Complexity
- **O(n)** for concatenated strings

```python
from typing import List

class Solution:
    def arrayStringsAreEqual(self, word1: List[str], word2: List[str]) -> bool:
        return ''.join(word1) == ''.join(word2)
```

# Test Cases
---

```python
def test_array_strings_are_equal():
    sol = Solution()

    assert sol.arrayStringsAreEqual(["ab", "c"], ["a", "bc"]) == True
    assert sol.arrayStringsAreEqual(["a", "cb"], ["ab", "c"]) == False
    assert sol.arrayStringsAreEqual(["abc", "d", "defg"], ["abcddefg"]) == True

    print("All test cases passed!")

test_array_strings_are_equal()
```
