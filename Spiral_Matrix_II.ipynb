# Spiral Matrix II

**Problem Link:**  
https://leetcode.com/problems/spiral-matrix-ii/description/

## Problem Statement
Given a positive integer `n`, generate an `n x n` matrix filled with elements from `1` to `n²` in spiral order.

## Examples

### Example 1
Input: `n = 3`  
Output:
```
[[1,2,3],
 [8,9,4],
 [7,6,5]]
```

### Example 2
Input: `n = 1`  
Output: `[[1]]`

## Constraints
- `1 <= n <= 20`


## Shrinking Boundary Fill

### Strategy
This is the inverse of Spiral Matrix (54) — instead of reading values out in spiral order, we write values `1..n²` into a matrix in spiral order using the same shrinking boundary technique.

- Initialise an `n × n` matrix of zeros
- Maintain four boundaries: `top`, `bottom`, `left`, `right`
- Fill one side at a time, then shrink the corresponding boundary:
  1. **Left → Right** along `top`, then `top += 1`
  2. **Top → Bottom** along `right`, then `right -= 1`
  3. **Right → Left** along `bottom`, then `bottom -= 1`
  4. **Bottom → Top** along `left`, then `left += 1`
- Repeat until `num > n²`
- No guard checks needed (unlike Spiral Matrix I read) because the loop condition `num <= n*n` stops exactly when all cells are filled

### Time Complexity
- **O(n²)** — every cell is written exactly once

### Space Complexity
- **O(n²)** for the output matrix (O(1) extra beyond the output)


````python
from typing import List

class Solution:
    def generateMatrix(self, n: int) -> List[List[int]]:
        matrix = [[0] * n for _ in range(n)]
        top, bottom = 0, n - 1
        left, right = 0, n - 1
        num = 1

        while num <= n * n:
            # 1. Left → Right along top row
            for col in range(left, right + 1):
                matrix[top][col] = num
                num += 1
            top += 1

            # 2. Top → Bottom along right column
            for row in range(top, bottom + 1):
                matrix[row][right] = num
                num += 1
            right -= 1

            # 3. Right → Left along bottom row
            for col in range(right, left - 1, -1):
                matrix[bottom][col] = num
                num += 1
            bottom -= 1

            # 4. Bottom → Top along left column
            for row in range(bottom, top - 1, -1):
                matrix[row][left] = num
                num += 1
            left += 1

        return matrix
````

````python
def test_generate_matrix():
    sol = Solution()

    assert sol.generateMatrix(3) == [[1,2,3],[8,9,4],[7,6,5]]
    assert sol.generateMatrix(1) == [[1]]

    # n = 2
    assert sol.generateMatrix(2) == [[1,2],[4,3]]

    # n = 4
    assert sol.generateMatrix(4) == [
        [ 1,  2,  3,  4],
        [12, 13, 14,  5],
        [11, 16, 15,  6],
        [10,  9,  8,  7]
    ]

    # All values 1..n² present exactly once
    for n in range(1, 6):
        matrix = sol.generateMatrix(n)
        flat   = [matrix[r][c] for r in range(n) for c in range(n)]
        assert sorted(flat) == list(range(1, n * n + 1)), f"Missing values for n={n}"

    print("All test cases passed!")

test_generate_matrix()
````
