# Rearrange Spaces Between Words

**Problem Link:**  
https://leetcode.com/problems/rearrange-spaces-between-words/

---

## Problem Statement

Given a string `text` containing words and spaces, redistribute the spaces so that:
- The number of spaces between adjacent words is equal and maximized.
- Any leftover spaces are placed at the end.
- The final string length remains the same.

---

## Examples

### Example 1
Input: "  this   is  a sentence "  
Output: "this   is   a   sentence"

### Example 2
Input: " practice   makes   perfect"  
Output: "practice   makes   perfect "

# Approach
---

## Counting Spaces + Even Distribution

### Strategy
- Count total spaces in the string
- Extract words using `split()`
- If only one word → all spaces go to the end
- Otherwise:
  - Even spaces = total_spaces // (number_of_words - 1)
  - Extra spaces = total_spaces % (number_of_words - 1)
- Join words using even spaces
- Append extra spaces at the end

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)**

```python
class Solution:
    def reorderSpaces(self, text: str) -> str:
        total_spaces = text.count(' ')
        words = text.split()

        if len(words) == 1:
            return words[0] + ' ' * total_spaces

        even_spaces = total_spaces // (len(words) - 1)
        extra_spaces = total_spaces % (len(words) - 1)

        return (' ' * even_spaces).join(words) + ' ' * extra_spaces
```

# Test Cases
---

```python
def test_reorder_spaces():
    sol = Solution()

    assert sol.reorderSpaces("  this   is  a sentence ") == "this   is   a   sentence"
    assert sol.reorderSpaces(" practice   makes   perfect") == "practice   makes   perfect "

    print("All test cases passed!")

test_reorder_spaces()
```
