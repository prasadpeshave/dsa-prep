# Word Search

## Problem Link
https://leetcode.com/problems/word-search/

---

## Problem Statement
Given a 2D board and a word, return True if the word exists in the grid.

- You can move **up, down, left, right**
- Each cell can be used **only once per path**

## Approach

### Backtracking (DFS)

- Try starting from every cell
- Perform DFS to match characters one by one
- Mark visited cells temporarily
- Backtrack after exploring

### Key Idea
Explore all paths but prune early if mismatch occurs.

### Time Complexity
- O(m * n * 4^L)

### Space Complexity
- O(L) recursion stack

```python
from typing import List

class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        m, n = len(board), len(board[0])

        def dfs(i, j, k):
            if k == len(word):
                return True

            if i < 0 or i >= m or j < 0 or j >= n or board[i][j] != word[k]:
                return False

            temp = board[i][j]
            board[i][j] = '#'

            found = (
                dfs(i+1, j, k+1) or
                dfs(i-1, j, k+1) or
                dfs(i, j+1, k+1) or
                dfs(i, j-1, k+1)
            )

            board[i][j] = temp
            return found

        for i in range(m):
            for j in range(n):
                if dfs(i, j, 0):
                    return True

        return False
```

## Test Cases

```python
def test_word_search():
    sol = Solution()

    board1 = [
        ['A','B','C','E'],
        ['S','F','C','S'],
        ['A','D','E','E']
    ]

    assert sol.exist([row[:] for row in board1], 'ABCCED') == True
    assert sol.exist([row[:] for row in board1], 'SEE') == True
    assert sol.exist([row[:] for row in board1], 'ABCB') == False

    print('All tests passed!')

test_word_search()
```
