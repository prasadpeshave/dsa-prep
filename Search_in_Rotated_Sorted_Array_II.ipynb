## Problem Statement
There is an integer array `nums` sorted in non-decreasing order (possibly with **duplicates**). Before being passed to your function, `nums` is rotated at an unknown pivot index `k`.

Given the array `nums` after the rotation and an integer `target`, return `true` if `target` is in `nums`, or `false` otherwise.

## Examples

### Example 1
Input: `nums = [2,5,6,0,0,1,2], target = 0`  
Output: `true`

### Example 2
Input: `nums = [2,5,6,0,0,1,2], target = 3`  
Output: `false`

## Constraints
- `1 <= nums.length <= 5000`
- `-10⁴ <= nums[i] <= 10⁴`
- `nums` is guaranteed to be rotated at some pivot
- `-10⁴ <= target <= 104`

## Modified Binary Search with Duplicate Handling

### Difference from Problem 33
In problem 33 (all distinct), comparing `nums[left]` with `nums[mid]` always tells us which half is sorted. With duplicates, `nums[left] == nums[mid]` creates ambiguity — we can't determine which side is sorted.

**Example:** `[1,3,1,1,1]` with `mid` pointing to another `1` — we can't tell if the left or right half is sorted.

### Strategy
- When `nums[left] == nums[mid]` → we can't determine which half is sorted. Safely increment `left` by 1 to skip the duplicate and try again
- When `nums[left] < nums[mid]` → left half is sorted (same logic as problem 33)
- When `nums[left] > nums[mid]` → right half is sorted (same logic as problem 33)

### Impact of Duplicates on Complexity
- **Best/Average case:** O(log n) — binary search still works most of the time
- **Worst case: O(n)** — when all elements are duplicates (e.g. `[1,1,1,1,1]`), we fall back to linear scanning one element at a time

### Space Complexity
- **O(1)** extra space


```python
from typing import List

class Solution:
    def search(self, nums: List[int], target: int) -> bool:
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return True

            # Duplicate ambiguity: can't determine which half is sorted
            if nums[left] == nums[mid]:
                left += 1
                continue

            # Left half is sorted
            if nums[left] < nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            # Right half is sorted
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1

        return False
```

```python
def test_search():
    sol = Solution()

    assert sol.search([2,5,6,0,0,1,2], 0)  == True
    assert sol.search([2,5,6,0,0,1,2], 3)  == False

    # All duplicates — worst case O(n)
    assert sol.search([1,1,1,1,1], 1)       == True
    assert sol.search([1,1,1,1,1], 2)       == False

    # No rotation
    assert sol.search([1,2,3,4,5], 3)       == True
    assert sol.search([1,2,3,4,5], 6)       == False

    # Target is the pivot element
    assert sol.search([4,5,6,7,0,1,2], 4)   == True

    # Single element
    assert sol.search([1], 1)               == True
    assert sol.search([1], 0)               == False

    # Duplicates spanning the rotation point
    assert sol.search([1,3,1,1,1], 3)       == True
    assert sol.search([1,3,1,1,1], 2)       == False

    # Target at boundary
    assert sol.search([3,1,1], 3)           == True
    assert sol.search([1,1,3], 3)           == True

    print("All test cases passed!")

test_search()
```
