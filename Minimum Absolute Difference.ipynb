# Minimum Absolute Difference

## Problem Link
https://leetcode.com/problems/minimum-absolute-difference/

---

## Problem Statement

Given an array of distinct integers `arr`, find all pairs of elements with the minimum absolute difference.

Each pair [a, b] must satisfy:
- a < b
- b - a equals the minimum absolute difference

Return pairs in ascending order.

## Approach

### Sorting + Adjacent Comparison
- Sort the array
- Compute minimum difference between adjacent elements
- Collect all pairs having that difference

### Why Sorting?
The smallest absolute difference must occur between adjacent elements in sorted order.

### Time Complexity
- O(n log n)

### Space Complexity
- O(1) (excluding output)

```python
from typing import List

class Solution:
    def minimumAbsDifference(self, arr: List[int]) -> List[List[int]]:
        arr.sort()
        min_diff = float('inf')

        # First pass to find minimum difference
        for i in range(1, len(arr)):
            min_diff = min(min_diff, arr[i] - arr[i - 1])

        # Second pass to collect pairs
        result = []
        for i in range(1, len(arr)):
            if arr[i] - arr[i - 1] == min_diff:
                result.append([arr[i - 1], arr[i]])

        return result
```

## Test Cases

```python
def test_minimum_abs_difference():
    sol = Solution()

    assert sol.minimumAbsDifference([4,2,1,3]) == [[1,2],[2,3],[3,4]]
    assert sol.minimumAbsDifference([1,3,6,10,15]) == [[1,3]]
    assert sol.minimumAbsDifference([3,8,-10,23,19,-4,-14,27]) == [[-14,-10],[19,23],[23,27]]

    print("All test cases passed!")

test_minimum_abs_difference()
```
