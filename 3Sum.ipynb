# 3Sum

**Problem Link:**  
https://leetcode.com/problems/3sum/description/

## Problem Statement
Given an integer array `nums`, return all the triplets `[nums[i], nums[j], nums[k]]` such that `i != j`, `i != k`, `j != k`, and `nums[i] + nums[j] + nums[k] == 0`.

The solution set must **not** contain duplicate triplets.

## Examples

### Example 1
Input: `nums = [-1,0,1,2,-1,-4]`  
Output: `[[-1,-1,2],[-1,0,1]]`  
Explanation: The distinct triplets that sum to zero are `[-1,-1,2]` and `[-1,0,1]`. Order of output and order within triplets does not matter.

### Example 2
Input: `nums = [0,1,1]`  
Output: `[]`  
Explanation: The only possible triplet does not sum to 0.

### Example 3
Input: `nums = [0,0,0]`  
Output: `[[0,0,0]]`

## Constraints
- `3 <= nums.length <= 3000`
- `-10⁵ <= nums[i] <= 10⁵`


## Sort + Two Pointers

### Strategy
- **Sort** `nums` — this enables the two-pointer technique and makes duplicate skipping straightforward
- Fix the first element `nums[i]` with an outer loop over indices `0` to `n-3`
  - Skip duplicate values of `nums[i]` to avoid duplicate triplets
  - Early exit: if `nums[i] > 0`, no triplet can sum to zero (all remaining values are ≥ `nums[i]`)
- For each fixed `i`, use two pointers `left = i+1` and `right = n-1` to find pairs that sum to `-nums[i]`:
  - If `nums[i] + nums[left] + nums[right] == 0` → record the triplet, then skip duplicate values of `left` and `right` before moving both pointers inward
  - If the sum is too small → move `left` right
  - If the sum is too large → move `right` left

### Time Complexity
- **O(n²)** — sorting is O(n log n), but the outer loop with inner two-pointer scan is O(n²)

### Space Complexity
- **O(1)** extra space (excluding the output list); sorting is in-place


```python
from typing import List

class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        result = []
        n = len(nums)

        for i in range(n - 2):
            # Early exit: smallest remaining value is positive, no zero-sum possible
            if nums[i] > 0:
                break

            # Skip duplicate values for the first element
            if i > 0 and nums[i] == nums[i - 1]:
                continue

            left, right = i + 1, n - 1

            while left < right:
                total = nums[i] + nums[left] + nums[right]

                if total == 0:
                    result.append([nums[i], nums[left], nums[right]])
                    # Skip duplicates for left and right
                    while left < right and nums[left] == nums[left + 1]:
                        left += 1
                    while left < right and nums[right] == nums[right - 1]:
                        right -= 1
                    left  += 1
                    right -= 1
                elif total < 0:
                    left += 1
                else:
                    right -= 1

        return result
```

```python
def test_three_sum():
    sol = Solution()

    # Helper: compare results regardless of order
    def normalize(triplets):
        return sorted([sorted(t) for t in triplets])

    assert normalize(sol.threeSum([-1, 0, 1, 2, -1, -4])) == normalize([[-1, -1, 2], [-1, 0, 1]])
    assert normalize(sol.threeSum([0, 1, 1]))              == []
    assert normalize(sol.threeSum([0, 0, 0]))              == [[0, 0, 0]]

    # All positive — no triplet sums to zero
    assert normalize(sol.threeSum([1, 2, 3]))              == []

    # All negative — no triplet sums to zero
    assert normalize(sol.threeSum([-3, -2, -1]))           == []

    # Multiple duplicates
    assert normalize(sol.threeSum([-2, 0, 0, 2, 2]))       == normalize([[-2, 0, 2]])

    # Larger input with multiple valid triplets
    assert normalize(sol.threeSum([-4, -1, -1, 0, 1, 2])) == normalize([[-1, -1, 2], [-1, 0, 1]])

    print("All test cases passed!")

test_three_sum()
```
