# Combinations

**Problem Link:**  
https://leetcode.com/problems/combinations/description/

## Problem Statement
Given two integers `n` and `k`, return all possible combinations of `k` numbers chosen from the range `[1, n]`.

You may return the answer in any order. Combinations are unordered — `[1,2]` and `[2,1]` are the same combination.

## Examples

### Example 1
Input: `n = 4, k = 2`  
Output: `[[1,2],[1,3],[1,4],[2,3],[2,4],[3,4]]`  
Explanation: There are C(4,2) = 6 total combinations.

### Example 2
Input: `n = 1, k = 1`  
Output: `[[1]]`  
Explanation: There is C(1,1) = 1 total combination.

## Constraints
- `1 <= n <= 20`
- `1 <= k <= n`


## Backtracking with Pruning

### Strategy
- Use a recursive `backtrack(start, current)` function:
  - `start` — the smallest number we may add next (ensures combinations are always in increasing order, preventing duplicates like `[2,1]` alongside `[1,2]`)
  - `current` — the combination being built
- **Base case:** when `len(current) == k`, record a copy and return
- **Recursive case:** iterate `i` from `start` to `n` inclusive, append `i`, recurse, then pop (backtrack)

### Pruning Optimisation
We can avoid exploring branches that can never yield a full combination of size `k`:
- At any point, if the numbers remaining `(n - i + 1)` are fewer than the numbers still needed `(k - len(current))`, no valid combination is possible from this branch
- This limits the upper bound of the loop to `n - (k - len(current)) + 1`

### Time Complexity
- **O(C(n, k) × k)** — C(n, k) combinations, each of length `k` to build and copy

### Space Complexity
- **O(k)** for the recursion call stack depth and `current` list


```python
from typing import List

class Solution:
    def combine(self, n: int, k: int) -> List[List[int]]:
        result = []

        def backtrack(start: int, current: list):
            if len(current) == k:
                result.append(list(current))
                return
            # Pruning: only iterate as far as there are enough numbers left
            need  = k - len(current)
            limit = n - need + 1
            for i in range(start, limit + 1):
                current.append(i)
                backtrack(i + 1, current)
                current.pop()

        backtrack(1, [])
        return result
```

```python
from math import comb

def test_combine():
    sol = Solution()

    def normalize(combos):
        return sorted([sorted(c) for c in combos])

    # Example 1
    assert normalize(sol.combine(4, 2)) == normalize([[1,2],[1,3],[1,4],[2,3],[2,4],[3,4]])

    # Example 2
    assert normalize(sol.combine(1, 1)) == [[1]]

    # k = 1 — each number is its own combination
    assert normalize(sol.combine(4, 1)) == [[1],[2],[3],[4]]

    # k = n — only one combination: all numbers
    assert normalize(sol.combine(4, 4)) == [[1,2,3,4]]

    # Count check: C(n, k) combinations for each valid pair
    for n in range(1, 8):
        for k in range(1, n + 1):
            result   = sol.combine(n, k)
            expected = comb(n, k)
            assert len(result) == expected, f"C({n},{k}): expected {expected}, got {len(result)}"

    # No duplicate combinations in output
    result = sol.combine(5, 3)
    normalized = normalize(result)
    assert len(normalized) == len(set(map(tuple, normalized))), "Duplicate combinations found"

    # Every combination is strictly increasing (no repeats within a combo)
    for combo in sol.combine(5, 3):
        assert combo == sorted(combo) and len(combo) == len(set(combo))

    print("All test cases passed!")

test_combine()
```
