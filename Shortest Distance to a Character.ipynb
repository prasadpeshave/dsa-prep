# Shortest Distance to a Character

**Problem Link:**  
https://leetcode.com/problems/shortest-distance-to-a-character/

---

## Problem Statement

Given a string `s` and a character `c` that occurs in `s`, return an array `answer` where `answer[i]` is the distance from index `i` to the closest occurrence of `c`.

Distance between two indices `i` and `j` is `abs(i - j)`.

## Example

**Input:** `s = "loveleetcode", c = "e"`

**Output:**
`[3,2,1,0,1,0,0,1,2,2,1,0]`

---

## Constraints

- `1 <= s.length <= 10^4`
- `c` appears at least once in `s`
- Only lowercase English letters

# Approach
---

## Two-Pass Technique

### Strategy
1. Traverse from left → right:
   - Track last seen position of `c`
   - Compute temporary distances
2. Traverse from right → left:
   - Update distances using nearest occurrence from right

This guarantees minimum distance from either side.

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)**

```python
from typing import List

class Solution:
    def shortestToChar(self, s: str, c: str) -> List[int]:
        n = len(s)
        result = [float('inf')] * n

        # Left to right pass
        prev = float('-inf')
        for i in range(n):
            if s[i] == c:
                prev = i
            result[i] = i - prev

        # Right to left pass
        prev = float('inf')
        for i in range(n - 1, -1, -1):
            if s[i] == c:
                prev = i
            result[i] = min(result[i], prev - i)

        return result
```

# Test Cases
---

```python
def test_shortest_to_char():
    sol = Solution()

    assert sol.shortestToChar("loveleetcode", "e") == \
        [3,2,1,0,1,0,0,1,2,2,1,0]

    assert sol.shortestToChar("aaab", "b") == [3,2,1,0]

    assert sol.shortestToChar("a", "a") == [0]

    print("All test cases passed!")

test_shortest_to_char()
```
