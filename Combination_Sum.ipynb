# Combination Sum

**Problem Link:**  
https://leetcode.com/problems/combination-sum/description/

## Problem Statement
Given an array of distinct integers `candidates` and a target integer `target`, return a list of all unique combinations of `candidates` where the chosen numbers sum to `target`. You may return the combinations in any order.

The same number may be chosen from `candidates` an **unlimited number of times**. Two combinations are unique if the frequency of at least one of the chosen numbers is different.

## Examples

### Example 1
Input: `candidates = [2,3,6,7], target = 7`  
Output: `[[2,2,3],[7]]`  
Explanation: `2+2+3=7` and `7=7` are the only two combinations.

### Example 2
Input: `candidates = [2,3,5], target = 8`  
Output: `[[2,2,2,2],[2,3,3],[3,5]]`

### Example 3
Input: `candidates = [2], target = 1`  
Output: `[]`

## Constraints
- `1 <= candidates.length <= 30`
- `2 <= candidates[i] <= 40`
- All elements of `candidates` are distinct
- `1 <= target <= 40`


## Backtracking

### Strategy
- Use a recursive `backtrack(start, current, remaining)` function:
  - `start` — the index in `candidates` to begin from (prevents duplicate combinations by never going backwards)
  - `current` — the combination being built
  - `remaining` — how much more we need to reach the target
- **Base cases:**
  - `remaining == 0` → valid combination found, add a copy to results
  - `remaining < 0` → overshot, prune this branch
- **Recursive case:** for each candidate from `start` onward, add it to `current` and recurse with `remaining - candidate`, then backtrack by popping it
- Starting from `start` (not `0`) at each level ensures we only consider each candidate at its position or later, naturally avoiding duplicate combinations while still allowing reuse of the same element

### Optimization
- Sorting `candidates` first allows an **early termination**: if `candidates[i] > remaining`, all subsequent candidates will also exceed it — break out of the loop immediately

### Time Complexity
- **O(n^(t/m))** where `n` is the number of candidates, `t` is the target, and `m` is the minimum candidate value — the depth of the recursion tree is at most `t/m` with `n` branches per level

### Space Complexity
- **O(t/m)** for the recursion call stack depth


```python
from typing import List

class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        candidates.sort()   # enables early termination
        result = []

        def backtrack(start: int, current: list, remaining: int):
            if remaining == 0:
                result.append(list(current))
                return
            for i in range(start, len(candidates)):
                if candidates[i] > remaining:
                    break               # sorted → no point checking further
                current.append(candidates[i])
                backtrack(i, current, remaining - candidates[i])  # i, not i+1: reuse allowed
                current.pop()

        backtrack(0, [], target)
        return result
```

```python
def test_combination_sum():
    sol = Solution()

    def normalize(combos):
        return sorted([sorted(c) for c in combos])

    assert normalize(sol.combinationSum([2, 3, 6, 7], 7)) == normalize([[2, 2, 3], [7]])
    assert normalize(sol.combinationSum([2, 3, 5], 8))    == normalize([[2, 2, 2, 2], [2, 3, 3], [3, 5]])
    assert normalize(sol.combinationSum([2], 1))           == []

    # Target exactly equals one candidate
    assert normalize(sol.combinationSum([3, 5, 7], 7))    == normalize([[7]])

    # Only one candidate that divides evenly
    assert normalize(sol.combinationSum([3], 9))           == normalize([[3, 3, 3]])

    # No valid combination
    assert normalize(sol.combinationSum([5], 3))           == []

    # All combinations are unique (no duplicates)
    result = sol.combinationSum([2, 3, 6, 7], 7)
    normalized = normalize(result)
    assert len(normalized) == len(set(map(tuple, normalized))), "Duplicate combinations found"

    print("All test cases passed!")

test_combination_sum()
```
