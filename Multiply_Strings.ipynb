# Multiply Strings

**Problem Link:**  
https://leetcode.com/problems/multiply-strings/description/

## Problem Statement
Given two non-negative integers `num1` and `num2` represented as strings, return the product of `num1` and `num2`, also represented as a string.

> **Note:** You must not use any built-in BigInteger library or convert the inputs to integer directly.

## Examples

### Example 1
Input: `num1 = "2", num2 = "3"`  
Output: `"6"`

### Example 2
Input: `num1 = "123", num2 = "456"`  
Output: `"56088"`

## Constraints
- `1 <= num1.length, num2.length <= 200`
- `num1` and `num2` consist of digits only
- Both `num1` and `num2` do not contain any leading zero, except the number `0` itself


## Grade-School Multiplication with Position Mapping

### Strategy
Simulate the grade-school multiplication algorithm, but instead of building intermediate rows and summing them, accumulate products directly into a result array:

- The product of `num1` (length `m`) and `num2` (length `n`) has at most `m + n` digits
- Allocate a `pos` array of size `m + n` initialized to zeros
- For each digit pair `(i, j)` (iterating from the right), the product `d1 * d2` contributes to positions `i + j` (higher digit / carry) and `i + j + 1` (lower digit) in `pos`
- After processing all pairs, convert `pos` to a string, skipping any leading zeros

### Position Mapping
If `num1[i]` is at offset `i` from the left and `num2[j]` is at offset `j` from the left:
- The product lands at `pos[i+j]` (tens place) and `pos[i+j+1]` (units place)
- Carry from `pos[i+j+1]` propagates into `pos[i+j]`

### Time Complexity
- **O(m × n)** where `m` and `n` are the lengths of `num1` and `num2`

### Space Complexity
- **O(m + n)** for the result array


```python
class Solution:
    def multiply(self, num1: str, num2: str) -> str:
        if num1 == "0" or num2 == "0":
            return "0"

        m, n = len(num1), len(num2)
        pos  = [0] * (m + n)

        for i in range(m - 1, -1, -1):
            for j in range(n - 1, -1, -1):
                product = (ord(num1[i]) - ord('0')) * (ord(num2[j]) - ord('0'))
                p1, p2  = i + j, i + j + 1
                total   = product + pos[p2]

                pos[p2]  = total % 10
                pos[p1] += total // 10

        # Skip leading zeros and build result string
        result = "".join(str(d) for d in pos).lstrip("0")
        return result if result else "0"
```

```python
def test_multiply():
    sol = Solution()

    assert sol.multiply("2",   "3")    == "6"
    assert sol.multiply("123", "456")  == "56088"

    # Zero cases
    assert sol.multiply("0",   "0")    == "0"
    assert sol.multiply("0",   "999")  == "0"
    assert sol.multiply("999", "0")    == "0"

    # Single digits
    assert sol.multiply("9",   "9")    == "81"

    # One is "1"
    assert sol.multiply("1",   "999")  == "999"
    assert sol.multiply("999", "1")    == "999"

    # Large numbers
    assert sol.multiply("99",  "99")   == "9801"
    assert sol.multiply("999", "999")  == "998001"

    # Asymmetric lengths
    assert sol.multiply("12",  "345")  == "4140"

    print("All test cases passed!")

test_multiply()
```
