# Next Permutation

**Problem Link:**  
https://leetcode.com/problems/next-permutation/description/

## Problem Statement
A permutation of an array of integers is an arrangement of its members into a sequence or linear order.

The **next permutation** of an array of integers is the next lexicographically greater permutation. If no such arrangement is possible (the array is in descending order), rearrange it into the lowest possible order (ascending).

The replacement must be **in-place** and use only **constant extra memory**.

## Examples

### Example 1
Input: `nums = [1,2,3]`  
Output: `[1,3,2]`

### Example 2
Input: `nums = [3,2,1]`  
Output: `[1,2,3]`  
Explanation: No larger permutation exists — wrap around to the smallest.

### Example 3
Input: `nums = [1,1,5]`  
Output: `[1,5,1]`

## Constraints
- `1 <= nums.length <= 100`
- `0 <= nums[i] <= 100`


## Three-Step In-Place Algorithm

### Key Observation
To get the next permutation, we want to make the smallest possible increase at the rightmost possible position. The algorithm follows three steps:

**Step 1 — Find the pivot:**  
Scan from right to left to find the first index `i` where `nums[i] < nums[i + 1]`. This is the rightmost position where the array is not yet in descending order — the "pivot".
- If no such `i` exists, the entire array is descending (the largest permutation) → skip to Step 3 directly to reverse the whole array.

**Step 2 — Find the successor and swap:**  
Scan from right to left to find the smallest element larger than `nums[i]` (call its index `j`), then swap `nums[i]` and `nums[j]`.

**Step 3 — Reverse the suffix:**  
Reverse the subarray from `i + 1` to the end. Since this suffix was in descending order before the swap, reversing it makes it ascending — the smallest possible arrangement for those positions.

### Time Complexity
- **O(n)** — at most two scans and one reversal, all linear

### Space Complexity
- **O(1)** — fully in-place


```python
from typing import List

class Solution:
    def nextPermutation(self, nums: List[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)

        # Step 1: find the rightmost pivot where nums[i] < nums[i+1]
        i = n - 2
        while i >= 0 and nums[i] >= nums[i + 1]:
            i -= 1

        if i >= 0:
            # Step 2: find the smallest element to the right of i that is greater than nums[i]
            j = n - 1
            while nums[j] <= nums[i]:
                j -= 1
            nums[i], nums[j] = nums[j], nums[i]

        # Step 3: reverse the suffix starting at i+1
        left, right = i + 1, n - 1
        while left < right:
            nums[left], nums[right] = nums[right], nums[left]
            left  += 1
            right -= 1
```

```python
def test_next_permutation():
    sol = Solution()

    def check(nums, expected):
        sol.nextPermutation(nums)
        assert nums == expected, f"Expected {expected}, got {nums}"

    check([1, 2, 3],    [1, 3, 2])
    check([3, 2, 1],    [1, 2, 3])   # largest permutation → wraps to smallest
    check([1, 1, 5],    [1, 5, 1])

    # Single element — stays the same
    check([1],          [1])

    # Two elements ascending
    check([1, 2],       [2, 1])

    # Two elements descending — wraps
    check([2, 1],       [1, 2])

    # Larger example
    check([1, 3, 2],    [2, 1, 3])
    check([2, 3, 1],    [3, 1, 2])
    check([1, 2, 4, 3], [1, 3, 2, 4])

    # Duplicate values
    check([1, 5, 1],    [5, 1, 1])

    print("All test cases passed!")

test_next_permutation()
```
