# Unique Morse Code Words

**Problem Link:**  
https://leetcode.com/problems/unique-morse-code-words/description/

## Problem Statement
International Morse Code defines a standard encoding where each letter is mapped to a series of dots and dashes. The full table for the 26 letters of the English alphabet is:

```
[".-","-...","-.-.","-..",".","..-.","--.","....","..",".---","-.-",".-..","--",
 "-.","---",".--.","--.-",".-.","...","-","..-","...-",".--","-..-","-.--","--.."]
```

Given an array of strings `words`, each word can be written as a concatenation of the Morse code of each letter — called the **transformation** of that word.

Return the number of **different** transformations among all words.

## Examples

### Example 1
Input: `words = ["gin","zen","gig","msg"]`  
Output: `2`  
Explanation:
- `"gin"` → `"--...-."`
- `"zen"` → `"--...-."`
- `"gig"` → `"--...--."` 
- `"msg"` → `"--...--."`

There are 2 different transformations: `"--...-."` and `"--...--."` .

### Example 2
Input: `words = ["a"]`  
Output: `1`

## Constraints
- `1 <= words.length <= 100`
- `1 <= words[i].length <= 12`
- `words[i]` consists of lowercase English letters


## Hash Set of Transformations

### Strategy
- Define the Morse code lookup table as a list of 26 strings indexed by `ord(ch) - ord('a')`
- For each word, concatenate the Morse codes of its letters to get its transformation string
- Add every transformation to a **set** — duplicates are automatically discarded
- Return the size of the set, which is the count of unique transformations

### Time Complexity
- **O(n × m)** where `n` is the number of words and `m` is the average word length

### Space Complexity
- **O(n × m)** for storing the transformation strings in the set


````python
from typing import List

class Solution:
    def uniqueMorseRepresentations(self, words: List[str]) -> int:
        morse = [
            ".-","-...","-.-.","-..",".","..-.","--.",
            "....","..",".---","-.-",".-..","--","-.",
            "---",".--.","--.-",".-.","...","-","..-",
            "...-",".--","-..-","-.--","--.."
        ]

        transformations = set()
        for word in words:
            code = "".join(morse[ord(ch) - ord('a')] for ch in word)
            transformations.add(code)

        return len(transformations)
````

````python
def test_unique_morse_representations():
    sol = Solution()

    # Example 1: "gin" and "zen" share a transformation, "gig" and "msg" share another
    assert sol.uniqueMorseRepresentations(["gin", "zen", "gig", "msg"]) == 2

    # Example 2: single word always gives 1 unique transformation
    assert sol.uniqueMorseRepresentations(["a"]) == 1

    # All identical words → 1 unique transformation
    assert sol.uniqueMorseRepresentations(["abc", "abc", "abc"]) == 1

    # All distinct transformations → count equals number of words
    assert sol.uniqueMorseRepresentations(["a", "b", "c"]) == 3

    # Single character words covering first five letters
    assert sol.uniqueMorseRepresentations(["a", "b", "c", "d", "e"]) == 5

    print("All test cases passed!")

test_unique_morse_representations()
````
