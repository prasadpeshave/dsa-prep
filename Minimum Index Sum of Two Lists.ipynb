# Minimum Index Sum of Two Lists

## Problem Link
https://leetcode.com/problems/minimum-index-sum-of-two-lists/

---

## Problem Statement

Find the common strings between two lists with the minimum index sum.
If a string appears at index i in list1 and j in list2, its index sum is i + j.
Return all common strings having the smallest index sum.

## Approach

### Hash Map + Single Pass
- Store indices of list1 elements in a dictionary
- Traverse list2 and check for common elements
- Compute index sum and track minimum
- Update result list accordingly

### Time Complexity
- O(n + m)

### Space Complexity
- O(n)

```python
from typing import List

class Solution:
    def findRestaurant(self, list1: List[str], list2: List[str]) -> List[str]:
        index_map = {value: i for i, value in enumerate(list1)}
        min_sum = float('inf')
        result = []

        for j, value in enumerate(list2):
            if value in index_map:
                current_sum = index_map[value] + j

                if current_sum < min_sum:
                    min_sum = current_sum
                    result = [value]
                elif current_sum == min_sum:
                    result.append(value)

        return result
```

## Test Cases

```python
def test_find_restaurant():
    sol = Solution()

    assert sol.findRestaurant(
        ["Shogun","Tapioca Express","Burger King","KFC"],
        ["Piatti","The Grill at Torrey Pines","Hungry Hunter Steakhouse","Shogun"]
    ) == ["Shogun"]

    assert sol.findRestaurant(
        ["Shogun","Tapioca Express","Burger King","KFC"],
        ["KFC","Shogun","Burger King"]
    ) == ["Shogun"]

    assert sorted(sol.findRestaurant(
        ["happy","sad","good"],
        ["sad","happy","good"]
    )) == ["happy","sad"]

    print("All test cases passed!")

test_find_restaurant()
```
