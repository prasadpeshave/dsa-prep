## Problem Statement

Given two non-negative integers `num1` and `num2` represented as strings, return their sum as a string.

### Restrictions
- Do **not** convert the entire string to an integer
- Do **not** use libraries for big integers

## Examples

### Example 1
**Input:** `num1 = "11", num2 = "123"`  
**Output:** `"134"`

---

### Example 2
**Input:** `num1 = "456", num2 = "77"`  
**Output:** `"533"`

---

### Example 3
**Input:** `num1 = "0", num2 = "0"`  
**Output:** `"0"`

---

## Constraints

- `1 <= num1.length, num2.length <= 10^4`
- Strings contain only digits
- No leading zeros except for zero itself

# Approach
---

## Digit-by-Digit Addition (Two Pointer)

### Strategy
- Start from the **end** of both strings
- Add digits along with a `carry`
- Store the result digit-by-digit
- Reverse the final result

### Why This Works
- Mimics manual addition
- Avoids integer conversion

### Time Complexity
- **O(n)** where `n = max(len(num1), len(num2))`

### Space Complexity
- **O(n)** for the result

```python
class Solution:
    def addStrings(self, num1: str, num2: str) -> str:
        i, j = len(num1) - 1, len(num2) - 1
        carry = 0
        result = []

        while i >= 0 or j >= 0 or carry:
            digit1 = ord(num1[i]) - ord('0') if i >= 0 else 0
            digit2 = ord(num2[j]) - ord('0') if j >= 0 else 0

            total = digit1 + digit2 + carry
            result.append(str(total % 10))
            carry = total // 10

            i -= 1
            j -= 1

        return ''.join(reversed(result))
```

# Test Cases
---

```python
def test_add_strings():
    sol = Solution()
    assert sol.addStrings("11", "123") == "134"
    assert sol.addStrings("456", "77") == "533"
    assert sol.addStrings("0", "0") == "0"
    assert sol.addStrings("999", "1") == "1000"
    print("All test cases passed!")

test_add_strings()
```
