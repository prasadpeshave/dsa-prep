# Add to Array-Form of Integer

**Problem Link:**  
https://leetcode.com/problems/add-to-array-form-of-integer/

---

## Problem Statement

The array-form of an integer represents its digits in left-to-right order.

Given an array `num` and an integer `k`, return the array-form of `num + k`.

## Examples

### Example 1
Input: `num = [1,2,0,0]`, `k = 34`  
Output: `[1,2,3,4]`

### Example 2
Input: `num = [2,7,4]`, `k = 181`  
Output: `[4,5,5]`

### Example 3
Input: `num = [2,1,5]`, `k = 806`  
Output: `[1,0,2,1]`

---

## Constraints

- `1 <= num.length <= 10^4`
- `0 <= num[i] <= 9`
- No leading zeros except zero itself
- `1 <= k <= 10^4`

# Approach
---

## Digit-by-Digit Addition (Manual Simulation)

### Strategy
- Start from the last digit of `num`
- Add `k` and propagate carry
- Continue until both `num` and `k` are exhausted
- Reverse the result at the end

### Time Complexity
- **O(n)**

### Space Complexity
- **O(n)**

```python
from typing import List

class Solution:
    def addToArrayForm(self, num: List[int], k: int) -> List[int]:
        result = []
        i = len(num) - 1

        while i >= 0 or k > 0:
            if i >= 0:
                k += num[i]
                i -= 1

            result.append(k % 10)
            k //= 10

        return result[::-1]
```

# Test Cases

```python
def test_add_to_array_form():
    sol = Solution()

    assert sol.addToArrayForm([1,2,0,0], 34) == [1,2,3,4]
    assert sol.addToArrayForm([2,7,4], 181) == [4,5,5]
    assert sol.addToArrayForm([2,1,5], 806) == [1,0,2,1]
    assert sol.addToArrayForm([0], 1000) == [1,0,0,0]

    print("All test cases passed!")

test_add_to_array_form()
```
