# Merge Intervals

**Problem Link:**  
https://leetcode.com/problems/merge-intervals/description/

## Problem Statement
Given an array of `intervals` where `intervals[i] = [start_i, end_i]`, merge all overlapping intervals and return an array of the non-overlapping intervals that cover all the intervals in the input.

## Examples

### Example 1
Input: `intervals = [[1,3],[2,6],[8,10],[15,18]]`  
Output: `[[1,6],[8,10],[15,18]]`  
Explanation: `[1,3]` and `[2,6]` overlap → merged into `[1,6]`.

### Example 2
Input: `intervals = [[1,4],[4,5]]`  
Output: `[[1,5]]`  
Explanation: Intervals touching at `4` are considered overlapping.

### Example 3
Input: `intervals = [[4,7],[1,4]]`  
Output: `[[1,7]]`  
Explanation: After sorting, `[1,4]` and `[4,7]` overlap and merge into `[1,7]`.

## Constraints
- `1 <= intervals.length <= 10⁴`
- `intervals[i].length == 2`
- `0 <= start_i <= end_i <= 10⁴`


## Sort + Linear Merge

### Strategy
- **Sort** intervals by their start value — this ensures overlapping intervals are always adjacent
- Initialize the result with the first interval as the current `merged` interval
- For each subsequent interval:
  - If its start is **≤** the current end → they overlap (or touch): extend the end to `max(current_end, interval_end)`
  - Otherwise → no overlap: append the current `merged` interval to results and start a new one
- Append the final `merged` interval after the loop

### Why sort first?
Without sorting, we'd need to compare every interval against every other — O(n²). Sorting guarantees that if interval `j` doesn't overlap with interval `i`, then no interval after `j` can overlap with `i` either (since all starts are ≥ `j`'s start).

### Time Complexity
- **O(n log n)** — dominated by sorting; the merge pass is O(n)

### Space Complexity
- **O(n)** for the output list (O(1) extra beyond output)


```python
from typing import List

class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort(key=lambda x: x[0])
        merged = [intervals[0]]

        for start, end in intervals[1:]:
            if start <= merged[-1][1]:          # overlapping or touching
                merged[-1][1] = max(merged[-1][1], end)
            else:                               # gap — start a new interval
                merged.append([start, end])

        return merged
```

```python
def test_merge():
    sol = Solution()

    assert sol.merge([[1,3],[2,6],[8,10],[15,18]])  == [[1,6],[8,10],[15,18]]
    assert sol.merge([[1,4],[4,5]])                 == [[1,5]]
    assert sol.merge([[4,7],[1,4]])                 == [[1,7]]

    # Single interval — no merge needed
    assert sol.merge([[1,5]])                       == [[1,5]]

    # All intervals already non-overlapping
    assert sol.merge([[1,2],[3,4],[5,6]])            == [[1,2],[3,4],[5,6]]

    # All intervals merge into one
    assert sol.merge([[1,4],[2,5],[3,6]])            == [[1,6]]

    # Nested interval (one fully inside another)
    assert sol.merge([[1,10],[2,5]])                 == [[1,10]]

    # Unsorted input
    assert sol.merge([[5,6],[1,2],[3,4]])            == [[1,2],[3,4],[5,6]]

    # Touching but not overlapping at boundary
    assert sol.merge([[1,3],[3,5]])                  == [[1,5]]

    print("All test cases passed!")

test_merge()
```
