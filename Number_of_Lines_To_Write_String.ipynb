# Number of Lines To Write String

**Problem Link:**  
https://leetcode.com/problems/number-of-lines-to-write-string/description/

## Problem Statement
You are given a string `s` of lowercase English letters and an array `widths` denoting how many pixels wide each lowercase English letter is. Specifically, `widths[0]` is the width of `'a'`, `widths[1]` is the width of `'b'`, and so on.

You are trying to write `s` across several lines, where each line is no longer than **100 pixels**. Starting at the beginning of `s`, write as many letters on the first line such that the total width does not exceed 100 pixels. Then continue writing as many letters as you can on the second line, and so on.

Return an array `result` of length 2 where:
- `result[0]` is the total number of lines
- `result[1]` is the width of the last line in pixels

## Examples

### Example 1
Input: `widths = [10,10,...,10]` (all 10), `s = "abcdefghijklmnopqrstuvwxyz"`  
Output: `[3, 60]`  
Explanation:
```
abcdefghij  → 100 pixels
klmnopqrst  → 100 pixels
uvwxyz      →  60 pixels
```

### Example 2
Input: `widths = [4,10,10,...,10]` (`'a'` is 4px, rest 10px), `s = "bbbcccdddaaa"`  
Output: `[2, 4]`  
Explanation:
```
bbbcccdddaa  →  98 pixels
a            →   4 pixels
```

## Constraints
- `widths.length == 26`
- `2 <= widths[i] <= 10`
- `1 <= s.length <= 1000`
- `s` contains only lowercase English letters


## Greedy Line Filling

### Strategy
- Track two variables: `lines` (number of lines used) and `current` (pixel width used on the current line)
- Start with `lines = 1` and `current = 0`
- For each character in `s`, look up its pixel width using `widths[ord(ch) - ord('a')]`
- If adding the character would exceed 100 pixels (`current + w > 100`), move to a new line: increment `lines` and reset `current = 0`
- Add the character's width to `current`
- Return `[lines, current]` at the end

### Time Complexity
- **O(n)** where `n` is the length of `s` — single pass

### Space Complexity
- **O(1)** extra space


````python
from typing import List

class Solution:
    def numberOfLines(self, widths: List[int], s: str) -> List[int]:
        lines   = 1
        current = 0

        for ch in s:
            w = widths[ord(ch) - ord('a')]
            if current + w > 100:
                lines   += 1
                current  = 0
            current += w

        return [lines, current]
````

````python
def test_number_of_lines():
    sol = Solution()

    # Example 1: all widths = 10, full alphabet
    widths_10 = [10] * 26
    assert sol.numberOfLines(widths_10, "abcdefghijklmnopqrstuvwxyz") == [3, 60]

    # Example 2: 'a' = 4px, rest = 10px
    widths_ex2 = [4] + [10] * 25
    assert sol.numberOfLines(widths_ex2, "bbbcccdddaaa") == [2, 4]

    # Single character that fits on one line
    assert sol.numberOfLines([10] * 26, "a") == [1, 10]

    # Character width exactly fills a line (100px)
    widths_exact = [10] * 26
    assert sol.numberOfLines(widths_exact, "aaaaaaaaaa") == [1, 100]  # 10 x 10 = 100

    # One character pushes onto a second line
    assert sol.numberOfLines(widths_exact, "aaaaaaaaaaa") == [2, 10]  # 11 x 10 → line 2

    # Minimum width letters filling many lines
    widths_min = [2] * 26
    # 50 letters per line, 100 letters → 2 lines, last line = 100px
    assert sol.numberOfLines(widths_min, "a" * 100) == [2, 100]

    print("All test cases passed!")

test_number_of_lines()
````
