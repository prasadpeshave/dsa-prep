# Gray Code

**Problem Link:**  
https://leetcode.com/problems/gray-code/description/

## Problem Statement
An **n-bit Gray code sequence** is a sequence of `2ⁿ` integers where:
- Every integer is in the inclusive range `[0, 2ⁿ - 1]`
- The first integer is `0`
- An integer appears no more than once in the sequence
- The binary representation of every pair of **adjacent** integers differs by exactly one bit
- The binary representation of the **first and last** integers also differs by exactly one bit

Given an integer `n`, return any valid n-bit Gray code sequence.

## Examples

### Example 1
Input: `n = 2`  
Output: `[0,1,3,2]`  
Explanation:
```
Binary: 00 → 01 → 11 → 10
Each adjacent pair differs by exactly 1 bit.
```

### Example 2
Input: `n = 1`  
Output: `[0,1]`

## Constraints
- `1 <= n <= 16`


## Mathematical Formula: i XOR (i >> 1)

### Key Insight
The standard Gray code for integer `i` is given by the formula:

```
gray(i) = i XOR (i >> 1)
```

This produces a valid Gray code sequence directly — no recursion or iteration needed beyond generating `i` from `0` to `2ⁿ - 1`.

### Why does this work?
Consider adjacent values `i` and `i+1`. They differ in their binary representation at the rightmost set bit of `i+1` and all lower bits. The XOR formula cancels the carry propagation, ensuring that `gray(i)` and `gray(i+1)` differ by exactly one bit.

**Example for n=2:**
```
i=0: 00 XOR 00 = 00 = 0
i=1: 01 XOR 00 = 01 = 1
i=2: 10 XOR 01 = 11 = 3
i=3: 11 XOR 01 = 10 = 2
→ [0, 1, 3, 2] ✓
```

### Alternative: Reflect-and-Prefix
A recursive construction: start with `[0]`, then for each bit, reflect the list and prepend `2^(bit-1)` to the reflected half. Both approaches produce equivalent sequences.

### Time Complexity
- **O(2ⁿ)** — generates all `2ⁿ` values

### Space Complexity
- **O(2ⁿ)** for the output list


````python
from typing import List

class Solution:
    def grayCode(self, n: int) -> List[int]:
        return [i ^ (i >> 1) for i in range(2 ** n)]
````

````python
def is_valid_gray_code(sequence, n):
    """Validate all Gray code properties."""
    expected_len = 2 ** n

    # Correct length
    if len(sequence) != expected_len:
        return False, f"Length {len(sequence)} != {expected_len}"

    # Starts with 0
    if sequence[0] != 0:
        return False, f"First element is {sequence[0]}, expected 0"

    # All values in [0, 2^n - 1] and unique
    if set(sequence) != set(range(expected_len)):
        return False, "Values are not a permutation of [0, 2^n - 1]"

    # Adjacent pairs (including wrap-around) differ by exactly one bit
    for i in range(len(sequence)):
        a = sequence[i]
        b = sequence[(i + 1) % len(sequence)]
        xor = a ^ b
        if xor == 0 or (xor & (xor - 1)) != 0:   # not a power of 2
            return False, f"Pair ({a}, {b}) differs by {bin(xor)}, not exactly 1 bit"

    return True, "Valid"

def test_gray_code():
    sol = Solution()

    # Known examples
    result = sol.grayCode(2)
    assert result == [0,1,3,2], f"Expected [0,1,3,2], got {result}"

    result = sol.grayCode(1)
    assert result == [0,1], f"Expected [0,1], got {result}"

    # Validate all properties for n = 1 to 8
    for n in range(1, 9):
        result = sol.grayCode(n)
        valid, msg = is_valid_gray_code(result, n)
        assert valid, f"n={n}: {msg}"

    print("All test cases passed!")

test_gray_code()
````
