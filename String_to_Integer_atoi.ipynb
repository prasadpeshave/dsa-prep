# String to Integer (atoi)

**Problem Link:**  
https://leetcode.com/problems/string-to-integer-atoi/description/

## Problem Statement
Implement the `myAtoi(string s)` function, which converts a string to a 32-bit signed integer.

The algorithm for `myAtoi(string s)` is as follows:
1. **Whitespace:** Ignore any leading whitespace (`' '`).
2. **Signedness:** Determine the sign by checking if the next character is `'-'` or `'+'`, assuming positivity if neither is present.
3. **Conversion:** Read the integer by skipping leading zeros until a non-digit character is encountered or the end of the string is reached. If no digits were read, the result is `0`.
4. **Rounding:** If the integer is out of the 32-bit signed integer range `[-2³¹, 2³¹ - 1]`, round to the nearest boundary.

## Examples

### Example 1
Input: `s = "42"`  
Output: `42`

### Example 2
Input: `s = " -042"`  
Output: `-42`  
Explanation: Leading whitespace ignored, `'-'` sets sign, leading zero ignored.

### Example 3
Input: `s = "1337c0d3"`  
Output: `1337`  
Explanation: Reading stops at the non-digit `'c'`.

### Example 4
Input: `s = "0-1"`  
Output: `0`  
Explanation: `'0'` is read, then reading stops at `'-'`.

### Example 5
Input: `s = "words and 987"`  
Output: `0`  
Explanation: Reading stops immediately at the non-digit `'w'`.

## Constraints
- `0 <= s.length <= 200`
- `s` consists of English letters, digits (`0-9`), `' '`, `'+'`, `'-'`, and `'.'`


## Step-by-Step Parsing

### Strategy
Follow the four steps of the algorithm directly using a pointer `i` into the string:

1. **Skip leading whitespace** — advance `i` while `s[i] == ' '`
2. **Read optional sign** — if `s[i]` is `'-'` or `'+'`, record the sign and advance `i`
3. **Read digits** — accumulate digits into `result` while `s[i].isdigit()`, then stop at any non-digit
4. **Clamp to 32-bit range** — apply `sign`, then clamp to `[-2³¹, 2³¹ - 1]`

### 32-bit Bounds
- `INT_MIN = -2³¹ = -2147483648`
- `INT_MAX =  2³¹ - 1 = 2147483647`

### Time Complexity
- **O(n)** — single pass through the string

### Space Complexity
- **O(1)** extra space


```python
class Solution:
    def myAtoi(self, s: str) -> int:
        INT_MIN = -(2 ** 31)
        INT_MAX =  (2 ** 31) - 1

        i    = 0
        n    = len(s)
        sign = 1
        result = 0

        # Step 1: skip leading whitespace
        while i < n and s[i] == ' ':
            i += 1

        # Step 2: read optional sign
        if i < n and s[i] in ('-', '+'):
            sign = -1 if s[i] == '-' else 1
            i += 1

        # Step 3: read digits
        while i < n and s[i].isdigit():
            result = result * 10 + int(s[i])
            i += 1

        # Step 4: apply sign and clamp
        result *= sign
        return max(INT_MIN, min(INT_MAX, result))
```

```python
def test_my_atoi():
    sol = Solution()

    # Examples from the problem
    assert sol.myAtoi("42")            == 42
    assert sol.myAtoi(" -042")         == -42
    assert sol.myAtoi("1337c0d3")      == 1337
    assert sol.myAtoi("0-1")           == 0
    assert sol.myAtoi("words and 987") == 0

    # Clamping to INT_MAX
    assert sol.myAtoi("99999999999")   == 2147483647

    # Clamping to INT_MIN
    assert sol.myAtoi("-99999999999")  == -2147483648

    # Exact boundary values
    assert sol.myAtoi("2147483647")    == 2147483647
    assert sol.myAtoi("-2147483648")   == -2147483648

    # Empty string
    assert sol.myAtoi("")              == 0

    # Only whitespace
    assert sol.myAtoi("   ")           == 0

    # Plus sign with number
    assert sol.myAtoi("+100")          == 100

    # Leading zeros
    assert sol.myAtoi("00042")         == 42

    print("All test cases passed!")

test_my_atoi()
```
