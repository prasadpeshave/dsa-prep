# Count and Say

**Problem Link:**  
https://leetcode.com/problems/count-and-say/description/

## Problem Statement
The count-and-say sequence is a sequence of digit strings defined by the recursive formula:
- `countAndSay(1) = "1"`
- `countAndSay(n)` is the run-length encoding (RLE) of `countAndSay(n - 1)`

**Run-length encoding (RLE):** Replace consecutive identical characters with the count followed by the character. For example, `"3322251"` → `"23321511"`.

Given a positive integer `n`, return the `n`th element of the count-and-say sequence.

## Examples

### Example 1
Input: `n = 4`  
Output: `"1211"`  
Explanation:
```
countAndSay(1) = "1"
countAndSay(2) = RLE of "1"    = "11"
countAndSay(3) = RLE of "11"   = "21"
countAndSay(4) = RLE of "21"   = "1211"
```

### Example 2
Input: `n = 1`  
Output: `"1"`  
Explanation: Base case.

## Constraints
- `1 <= n <= 30`


## Iterative RLE Generation

### Strategy
- Start with the base string `"1"`
- Repeat `n - 1` times: apply one round of run-length encoding to the current string to produce the next
- To encode a string:
  - Walk through it tracking the current character and its consecutive count
  - Whenever the character changes (or the end is reached), append `str(count) + char` to the result
  - Reset the count and move to the next character

### Time Complexity
- **O(n × m)** where `m` is the length of the string at step `n` — each of the `n` steps scans the current string once

### Space Complexity
- **O(m)** for the current and next strings


````python
class Solution:
    def countAndSay(self, n: int) -> str:
        current = "1"

        for _ in range(n - 1):
            next_str = []
            i = 0
            while i < len(current):
                ch    = current[i]
                count = 0
                while i < len(current) and current[i] == ch:
                    count += 1
                    i     += 1
                next_str.append(str(count) + ch)
            current = "".join(next_str)

        return current
````

````python
def test_count_and_say():
    sol = Solution()

    # Known sequence values
    assert sol.countAndSay(1)  == "1"
    assert sol.countAndSay(2)  == "11"
    assert sol.countAndSay(3)  == "21"
    assert sol.countAndSay(4)  == "1211"
    assert sol.countAndSay(5)  == "111221"
    assert sol.countAndSay(6)  == "312211"
    assert sol.countAndSay(7)  == "13112221"
    assert sol.countAndSay(8)  == "1113213211"

    # Each term is the RLE of the previous — verify the chain property
    for i in range(2, 11):
        prev = sol.countAndSay(i - 1)
        curr = sol.countAndSay(i)
        # Decode curr back and compare with prev
        decoded = ""
        j = 0
        while j < len(curr):
            count = int(curr[j])
            ch    = curr[j + 1]
            decoded += ch * count
            j += 2
        assert decoded == prev, f"RLE chain broken at n={i}"

    print("All test cases passed!")

test_count_and_say()
````
