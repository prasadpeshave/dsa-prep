# Convert Integer to the Sum of Two No‑Zero Integers

## Problem Link
https://leetcode.com/problems/convert-integer-to-the-sum-of-two-no-zero-integers/

---

## Problem Statement
A **No‑Zero integer** is a positive integer that does not contain the digit `0`.

Given an integer `n`, return two integers `[a, b]` such that:
- `a + b = n`
- both `a` and `b` do **not contain the digit `0`**.

There is guaranteed to be at least one valid solution.

## Approach

### Brute Force Search
- Iterate `a` from `1` to `n`
- Compute `b = n - a`
- Check whether `a` and `b` contain the digit `0`
- If both are valid No‑Zero integers, return `[a, b]`

Since `n ≤ 10⁴`, this simple search works efficiently.

### Time Complexity
- **O(n * d)** where `d` is number of digits

### Space Complexity
- **O(1)**

```python
from typing import List

class Solution:
    def getNoZeroIntegers(self, n: int) -> List[int]:
        for a in range(1, n):
            b = n - a
            if '0' not in str(a) and '0' not in str(b):
                return [a, b]
```

## Test Cases

```python
def test_get_no_zero_integers():
    sol = Solution()

    assert sol.getNoZeroIntegers(2) == [1,1]
    a,b = sol.getNoZeroIntegers(11)
    assert a + b == 11
    assert '0' not in str(a) and '0' not in str(b)

    print('All tests passed!')

test_get_no_zero_integers()
```
