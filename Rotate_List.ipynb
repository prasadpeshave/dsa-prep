# Rotate List

**Problem Link:**  
https://leetcode.com/problems/rotate-list/description/

## Problem Statement
Given the `head` of a linked list, rotate the list to the right by `k` places.

## Examples

### Example 1
Input: `head = [1,2,3,4,5], k = 2`  
Output: `[4,5,1,2,3]`

### Example 2
Input: `head = [0,1,2], k = 4`  
Output: `[2,0,1]`

## Constraints
- The number of nodes in the list is in the range `[0, 500]`
- `-100 <= Node.val <= 100`
- `0 <= k <= 2 × 10⁹`


## Form a Ring, Then Break It

### Strategy
Rotating right by `k` positions moves the last `k % n` nodes to the front.

1. **Find the length** by traversing to the tail, then connect `tail.next = head` to form a circular list
2. **Compute the effective rotation**: `k = k % n` (rotating by `n` returns the original list)
3. **Find the new tail**: it sits at position `n - k - 1` from the current head (0-indexed)
4. **Break the circle**: the node after the new tail becomes the new head; set `new_tail.next = None`

### Edge Cases
- Empty list or single node — return `head` immediately
- `k % n == 0` — no rotation needed; break the ring and return the original head

### Time Complexity
- **O(n)** — at most two passes through the list

### Space Complexity
- **O(1)** — only pointer manipulation, no extra data structures


```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val  = val
        self.next = next

class Solution:
    def rotateRight(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        if not head or not head.next or k == 0:
            return head

        # Step 1: find length and tail
        length = 1
        tail   = head
        while tail.next:
            tail   = tail.next
            length += 1

        # Step 2: effective rotation (rotating by n is a no-op)
        k = k % length
        if k == 0:
            return head

        # Step 3: form a circle
        tail.next = head

        # Step 4: find the new tail (length - k - 1 steps from head)
        new_tail = head
        for _ in range(length - k - 1):
            new_tail = new_tail.next

        # Step 5: break the circle
        new_head       = new_tail.next
        new_tail.next  = None

        return new_head
```

```python
# --- Helper utilities ---
def build_list(values):
    if not values:
        return None
    head = ListNode(values[0])
    curr = head
    for v in values[1:]:
        curr.next = ListNode(v)
        curr      = curr.next
    return head

def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

def test_rotate_right():
    sol = Solution()

    assert to_list(sol.rotateRight(build_list([1,2,3,4,5]), 2)) == [4,5,1,2,3]
    assert to_list(sol.rotateRight(build_list([0,1,2]),     4)) == [2,0,1]

    # k = 0 — no rotation
    assert to_list(sol.rotateRight(build_list([1,2,3]),     0)) == [1,2,3]

    # k equals length — full rotation, same list
    assert to_list(sol.rotateRight(build_list([1,2,3]),     3)) == [1,2,3]

    # k > length
    assert to_list(sol.rotateRight(build_list([1,2,3]),     7)) == [3,1,2]  # 7 % 3 = 1

    # Single node
    assert to_list(sol.rotateRight(build_list([1]),         5)) == [1]

    # Empty list
    assert to_list(sol.rotateRight(None,                    3)) == []

    # Two nodes
    assert to_list(sol.rotateRight(build_list([1,2]),       1)) == [2,1]

    print("All test cases passed!")

test_rotate_right()
```
