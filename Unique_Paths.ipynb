# Unique Paths

**Problem Link:**  
https://leetcode.com/problems/unique-paths/description/

## Problem Statement
There is a robot on an `m x n` grid. The robot is initially located at the top-left corner (`grid[0][0]`) and tries to reach the bottom-right corner (`grid[m-1][n-1]`). The robot can only move **right** or **down** at any point.

Given the two integers `m` and `n`, return the number of possible unique paths.

## Examples

### Example 1
Input: `m = 3, n = 7`  
Output: `28`

### Example 2
Input: `m = 3, n = 2`  
Output: `3`  
Explanation: The three paths are:
1. Right → Down → Down
2. Down → Down → Right
3. Down → Right → Down

## Constraints
- `1 <= m, n <= 100`


## Dynamic Programming

### Strategy
Define `dp[i][j]` as the number of unique paths to reach cell `(i, j)`.

- **Base cases:** The entire top row and left column each have exactly `1` path (can only arrive by going straight right or straight down respectively)
- **Recurrence:** `dp[i][j] = dp[i-1][j] + dp[i][j-1]` — the robot can only arrive from the cell above or the cell to the left
- **Answer:** `dp[m-1][n-1]`

### Space Optimisation
Since each row only depends on the row above it, we can compress the `m × n` DP table down to a **single 1D array** of length `n`, updating it row by row. This reduces space from O(m × n) to O(n).

### Mathematical Alternative
The answer is also the combinatorial value C(m+n-2, n-1) — the number of ways to choose `n-1` rightward moves out of `m+n-2` total moves. The DP approach is shown here as it generalises better to variations like Unique Paths II.

### Time Complexity
- **O(m × n)** — fills the entire DP table

### Space Complexity
- **O(n)** — single row array (optimised from O(m × n))


```python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        dp = [1] * n   # base case: top row is all 1s

        for _ in range(1, m):        # iterate over remaining rows
            for j in range(1, n):    # leftmost cell of each row stays 1 (base case)
                dp[j] += dp[j - 1]  # dp[j] (from above) + dp[j-1] (from left)

        return dp[n - 1]
```

```python
from math import comb

def test_unique_paths():
    sol = Solution()

    assert sol.uniquePaths(3, 7) == 28
    assert sol.uniquePaths(3, 2) == 3

    # Single row or single column — only one path
    assert sol.uniquePaths(1, 1)   == 1
    assert sol.uniquePaths(1, 10)  == 1
    assert sol.uniquePaths(10, 1)  == 1

    # Symmetric: uniquePaths(m, n) == uniquePaths(n, m)
    assert sol.uniquePaths(3, 5) == sol.uniquePaths(5, 3)
    assert sol.uniquePaths(4, 6) == sol.uniquePaths(6, 4)

    # Cross-check against combinatorial formula C(m+n-2, n-1)
    for m in range(1, 8):
        for n in range(1, 8):
            expected = comb(m + n - 2, n - 1)
            assert sol.uniquePaths(m, n) == expected, \
                f"uniquePaths({m},{n}) = {sol.uniquePaths(m,n)}, expected {expected}"

    print("All test cases passed!")

test_unique_paths()
```
