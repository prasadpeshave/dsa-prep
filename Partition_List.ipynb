# Partition List

**Problem Link:**  
https://leetcode.com/problems/partition-list/description/

## Problem Statement
Given the `head` of a linked list and a value `x`, partition it such that all nodes **less than** `x` come before nodes **greater than or equal to** `x`.

You should preserve the original relative order of the nodes in each of the two partitions.

## Examples

### Example 1
Input: `head = [1,4,3,2,5,2], x = 3`  
Output: `[1,2,2,4,3,5]`

### Example 2
Input: `head = [2,1], x = 2`  
Output: `[1,2]`

## Constraints
- The number of nodes in the list is in the range `[0, 200]`
- `-100 <= Node.val <= 100`
- `-200 <= x <= 200`


## Two Dummy-Head Lists

### Strategy
Build two separate sub-lists in a single pass, then join them:

- **`less`** — collects all nodes with `val < x` (in original order)
- **`greater`** — collects all nodes with `val >= x` (in original order)

Each sub-list uses a **sentinel (dummy) head** so we never need to special-case the first node. Walk `curr` through the original list, appending each node to the appropriate sub-list tail.

After the loop:
1. Terminate the `greater` list: `greater_tail.next = None` (prevents a cycle if the last node pointed somewhere)
2. Join the two lists: `less_tail.next = greater.next` (the first real node of the `greater` list)
3. Return `less.next` (first real node of the `less` list)

### Why Two Sentinels?
Without sentinel nodes we'd need to track whether either sub-list is empty. Sentinels eliminate all that branching — the tails always have a valid node to append to.

### Time Complexity
- **O(n)** — single pass through the list

### Space Complexity
- **O(1)** — only two extra sentinel nodes; no new nodes created


```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val  = val
        self.next = next

class Solution:
    def partition(self, head: Optional[ListNode], x: int) -> Optional[ListNode]:
        less    = ListNode(0)   # sentinel for the "< x" partition
        greater = ListNode(0)   # sentinel for the ">= x" partition

        less_tail    = less
        greater_tail = greater
        curr         = head

        while curr:
            if curr.val < x:
                less_tail.next = curr
                less_tail      = less_tail.next
            else:
                greater_tail.next = curr
                greater_tail      = greater_tail.next
            curr = curr.next

        greater_tail.next  = None          # terminate the greater list
        less_tail.next     = greater.next  # join: less → greater

        return less.next
```

```python
# --- Helper utilities ---
def build_list(values):
    if not values:
        return None
    head = ListNode(values[0])
    curr = head
    for v in values[1:]:
        curr.next = ListNode(v)
        curr      = curr.next
    return head

def to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

def test_partition():
    sol = Solution()

    assert to_list(sol.partition(build_list([1,4,3,2,5,2]), 3)) == [1,2,2,4,3,5]
    assert to_list(sol.partition(build_list([2,1]),          2)) == [1,2]

    # Empty list
    assert to_list(sol.partition(None,                        0)) == []

    # Single node less than x
    assert to_list(sol.partition(build_list([1]),             3)) == [1]

    # Single node greater than x
    assert to_list(sol.partition(build_list([5]),             3)) == [5]

    # All nodes less than x — order preserved
    assert to_list(sol.partition(build_list([1,2,2]),         5)) == [1,2,2]

    # All nodes greater than or equal to x — order preserved
    assert to_list(sol.partition(build_list([3,4,5]),         2)) == [3,4,5]

    # Already partitioned — no change in order
    assert to_list(sol.partition(build_list([1,2,4,5]),       3)) == [1,2,4,5]

    # x equal to a node value — that node goes to the >= partition
    assert to_list(sol.partition(build_list([3,1,2]),         3)) == [1,2,3]

    # Relative order preserved within each partition
    result = to_list(sol.partition(build_list([4,2,5,1,3]),   3))
    less_part    = [v for v in result if v < 3]
    greater_part = [v for v in result if v >= 3]
    assert less_part    == [2,1],   f"Less partition order wrong: {less_part}"
    assert greater_part == [4,5,3], f"Greater partition order wrong: {greater_part}"

    print("All test cases passed!")

test_partition()
```
