# Subsets

**Problem Link:**  
https://leetcode.com/problems/subsets/description/

## Problem Statement
Given an integer array `nums` of unique elements, return all possible subsets (the power set).

The solution set must not contain duplicate subsets. Return the solution in any order.

## Examples

### Example 1
Input: `nums = [1,2,3]`  
Output: `[[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]`

### Example 2
Input: `nums = [0]`  
Output: `[[],[0]]`

## Constraints
- `1 <= nums.length <= 10`
- `-10 <= nums[i] <= 10`
- All numbers in `nums` are unique


## Backtracking

### Strategy
- Use a recursive `backtrack(start, current)` function:
  - At **every** call (not just the base case), record a copy of `current` — this captures subsets of all sizes including the empty set
  - Iterate from `start` onward: append `nums[i]`, recurse with `i + 1`, then pop (backtrack)
  - Using `start` ensures elements are always picked in forward order, preventing duplicates like `[1,2]` and `[2,1]`

### How All Subset Sizes Are Captured
- When `backtrack(0, [])` is first called → `[]` is recorded (empty set)
- When `backtrack(1, [1])` is called → `[1]` is recorded
- When `backtrack(2, [1,2])` is called → `[1,2]` is recorded
- And so on — every node in the recursion tree is a valid subset

### Time Complexity
- **O(2ⁿ × n)** — 2ⁿ subsets, each up to length `n` to copy

### Space Complexity
- **O(n)** for the recursion call stack depth and `current` list (excluding the output)


```python
from typing import List

class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        result = []

        def backtrack(start: int, current: list):
            result.append(list(current))   # record subset at every node
            for i in range(start, len(nums)):
                current.append(nums[i])
                backtrack(i + 1, current)
                current.pop()

        backtrack(0, [])
        return result
```

```python
def test_subsets():
    sol = Solution()

    def normalize(subsets):
        return sorted([sorted(s) for s in subsets])

    # Example 1
    assert normalize(sol.subsets([1,2,3])) == normalize([[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]])

    # Example 2
    assert normalize(sol.subsets([0])) == normalize([[],[0]])

    # Single element — empty set and the element itself
    assert normalize(sol.subsets([5])) == normalize([[],[5]])

    # Count check: 2^n subsets for each input size
    for n in range(1, 8):
        nums   = list(range(1, n + 1))
        result = sol.subsets(nums)
        assert len(result) == 2 ** n, f"Expected {2**n} subsets for n={n}, got {len(result)}"

    # No duplicate subsets in output
    result     = sol.subsets([1,2,3])
    normalized = normalize(result)
    assert len(normalized) == len(set(map(tuple, normalized))), "Duplicate subsets found"

    # Empty set is always included
    assert [] in sol.subsets([1,2,3])

    # Full set is always included
    result = sol.subsets([1,2,3])
    assert sorted([1,2,3]) in [sorted(s) for s in result]

    print("All test cases passed!")

test_subsets()
```
