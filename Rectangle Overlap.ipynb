# Rectangle Overlap

## Problem Link
https://leetcode.com/problems/rectangle-overlap/

---

## Problem Statement
Two axis-aligned rectangles are represented as `[x1, y1, x2, y2]`, where `(x1, y1)` is the bottom-left corner and `(x2, y2)` is the top-right corner.

Return **True** if the rectangles overlap with **positive area**, otherwise return **False**.

Rectangles touching only at edges or corners are **not considered overlapping**.

## Approach

### Geometry Observation
Two rectangles **do not overlap** if one is completely on one side of the other.

Non-overlapping cases:
- One rectangle is completely to the **left** of the other
- One rectangle is completely to the **right** of the other
- One rectangle is completely **above** the other
- One rectangle is completely **below** the other

If none of these conditions hold, the rectangles overlap.

### Time Complexity
- **O(1)**

### Space Complexity
- **O(1)**

```python
from typing import List

class Solution:
    def isRectangleOverlap(self, rec1: List[int], rec2: List[int]) -> bool:
        # Check if one rectangle is completely left/right
        if rec1[2] <= rec2[0] or rec2[2] <= rec1[0]:
            return False

        # Check if one rectangle is completely above/below
        if rec1[3] <= rec2[1] or rec2[3] <= rec1[1]:
            return False

        return True
```

## Test Cases

```python
def test_rectangle_overlap():
    sol = Solution()

    assert sol.isRectangleOverlap([0,0,2,2], [1,1,3,3]) == True
    assert sol.isRectangleOverlap([0,0,1,1], [1,0,2,1]) == False
    assert sol.isRectangleOverlap([0,0,1,1], [2,2,3,3]) == False

    print('All test cases passed!')

test_rectangle_overlap()
```
