# Simplify Path

**Problem Link:**  
https://leetcode.com/problems/simplify-path/description/

## Problem Statement
You are given an absolute path for a Unix-style file system, which always begins with `'/'`. Transform it into its simplified canonical path.

**Unix-style file system rules:**
- `'.'` represents the current directory
- `'..'` represents the parent directory
- Multiple consecutive slashes (`'//'`, `'///'`) are treated as a single `'/'`
- Any other sequence of periods (e.g. `'...'`, `'....'`) is a valid directory/file name

**Canonical path rules:**
- Must start with `'/'`
- Directories separated by exactly one `'/'`
- No trailing `'/'` (unless it is the root)
- No `'.'` or `'..'` components

## Examples

### Example 1
Input: `"/home/"`  Output: `"/home"` — trailing slash removed

### Example 2
Input: `"/home//foo/"` Output: `"/home/foo"` — consecutive slashes collapsed

### Example 3
Input: `"/home/user/Documents/../Pictures"` Output: `"/home/user/Pictures"` — `..` goes up one level

### Example 4
Input: `"/../"` Output: `"/"` — can't go above root

### Example 5
Input: `"/.../a/../b/c/../d/./"` Output: `"/.../b/d"` — `'...'` is a valid name

## Constraints
- `1 <= path.length <= 3000`
- `path` consists of English letters, digits, `'.'`, `'/'`, or `'_'`
- `path` is a valid absolute Unix path


## Stack-Based Parsing

### Strategy
- Split `path` by `'/'` to get individual components (empty strings and `'.'` are skipped)
- Use a **stack** to simulate directory traversal:
  - `'..'` → pop from the stack (go up one level), if non-empty
  - `'.'` or `''` → skip (current directory or extra slash)
  - Any other name → push onto the stack (enter that directory)
- Join the stack with `'/'` and prepend `'/'` to form the canonical path
- If the stack is empty, return `'/'` (root directory)

### Why Split on `'/'`?
`path.split('/')` handles multiple consecutive slashes automatically — they produce empty strings which we simply skip. This eliminates the need to manually scan for slash sequences.

### Time Complexity
- **O(n)** where `n` is the length of `path` — one split and one linear scan through components

### Space Complexity
- **O(n)** for the stack and split components


```python
class Solution:
    def simplifyPath(self, path: str) -> str:
        stack = []

        for part in path.split('/'):
            if part == '..':
                if stack:
                    stack.pop()      # go up one level
            elif part and part != '.':
                stack.append(part)  # valid directory/file name
            # '' (extra slash) and '.' (current dir) are simply skipped

        return '/' + '/'.join(stack)
```

```python
def test_simplify_path():
    sol = Solution()

    assert sol.simplifyPath("/home/")                          == "/home"
    assert sol.simplifyPath("/home//foo/")                     == "/home/foo"
    assert sol.simplifyPath("/home/user/Documents/../Pictures") == "/home/user/Pictures"
    assert sol.simplifyPath("/../")                            == "/"
    assert sol.simplifyPath("/.../a/../b/c/../d/./")           == "/.../b/d"

    # Root directory
    assert sol.simplifyPath("/")                               == "/"

    # Current directory references
    assert sol.simplifyPath("/./././")                         == "/"
    assert sol.simplifyPath("/a/./b/./c")                      == "/a/b/c"

    # Multiple parent directory navigations
    assert sol.simplifyPath("/a/b/c/../../d")                  == "/a/d"
    assert sol.simplifyPath("/a/../../b")                      == "/b"

    # Three-dot directory name (valid!)
    assert sol.simplifyPath("/...")                            == "/..."

    # Already canonical
    assert sol.simplifyPath("/a/b/c")                          == "/a/b/c"

    print("All test cases passed!")

test_simplify_path()
```
