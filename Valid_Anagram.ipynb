# Valid Anagram

**Problem Link:**  
https://leetcode.com/problems/valid-anagram/description/

## Problem Statement
Given two strings `s` and `t`, return `true` if `t` is an anagram of `s`, and `false` otherwise.

## Examples

### Example 1
Input: `s = "anagram", t = "nagaram"`  
Output: `true`

### Example 2
Input: `s = "rat", t = "car"`  
Output: `false`

## Constraints
- `1 <= s.length, t.length <= 5 * 10^4`
- `s` and `t` consist of lowercase English letters

## Follow Up
What if the inputs contain Unicode characters? How would you adapt your solution to such a case?


## Character Frequency Count

### Strategy
- If lengths differ, they cannot be anagrams — return `False` early
- Use a hash map (dictionary) to count the frequency of each character in `s`
- Decrement the count for each character in `t`
- If any count is non-zero at the end, `t` is not an anagram of `s`

### Time Complexity
- **O(n)** where `n` is the length of the strings

### Space Complexity
- **O(1)** — at most 26 keys in the map for lowercase English letters (O(n) for Unicode follow-up)

### Follow Up — Unicode Characters
- The same hash map approach works directly for Unicode; Python's `dict` handles any hashable character
- Space becomes **O(n)** in the worst case since the character set is no longer bounded to 26


```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        count = {}

        for char in s:
            count[char] = count.get(char, 0) + 1

        for char in t:
            count[char] = count.get(char, 0) - 1
            if count[char] < 0:
                return False

        return True
```

```python
def test_is_anagram():
    sol = Solution()
    assert sol.isAnagram("anagram", "nagaram") == True
    assert sol.isAnagram("rat", "car") == False
    assert sol.isAnagram("a", "a") == True
    assert sol.isAnagram("ab", "a") == False
    assert sol.isAnagram("aacc", "ccac") == False
    assert sol.isAnagram("listen", "silent") == True
    print("All test cases passed!")

test_is_anagram()
```
