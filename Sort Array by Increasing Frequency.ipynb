# Sort Array by Increasing Frequency

## Problem Statement

Given an integer array `nums`, sort it in increasing order based on the frequency of values.

If multiple values have the same frequency, sort them in decreasing order.

---

## Examples

### Example 1
Input: [1,1,2,2,2,3]  
Output: [3,1,1,2,2,2]

### Example 2
Input: [2,3,1,3,2]  
Output: [1,3,3,2,2]

### Example 3
Input: [-1,1,-6,4,5,-6,1,4,1]  
Output: [5,-1,4,4,-6,-6,1,1,1]

# Approach
---

## Frequency + Custom Sorting

### Strategy
- Count frequency using Counter
- Sort using custom key:
  - Primary key → frequency (ascending)
  - Secondary key → value (descending)

We can achieve this using:
`key=lambda x: (freq[x], -x)`

### Time Complexity
- **O(n log n)**

### Space Complexity
- **O(n)**

```python
from typing import List
from collections import Counter

class Solution:
    def frequencySort(self, nums: List[int]) -> List[int]:
        freq = Counter(nums)
        return sorted(nums, key=lambda x: (freq[x], -x))
```

# Test Cases
---

```python
def test_frequency_sort():
    sol = Solution()

    assert sol.frequencySort([1,1,2,2,2,3]) == [3,1,1,2,2,2]
    assert sol.frequencySort([2,3,1,3,2]) == [1,3,3,2,2]
    assert sol.frequencySort([-1,1,-6,4,5,-6,1,4,1]) == [5,-1,4,4,-6,-6,1,1,1]

    print('All test cases passed!')

test_frequency_sort()
```
