## Problem Statement

The complement of an integer is obtained by flipping all bits in its binary representation (without leading zeros).

Given an integer `num`, return its complement.

---

## Examples

### Example 1
Input: num = 5  
Output: 2

Explanation: 5 -> 101, flipped -> 010 -> 2

### Example 2
Input: num = 1  
Output: 0

# Approach
---

## Bit Manipulation with Mask

### Strategy
- Find the bit-length of the number.
- Create a mask of all 1s with the same length.
- XOR the number with the mask.

Example:
num = 5 (101)
mask = 111
101 ^ 111 = 010

### Time Complexity
- **O(1)**

### Space Complexity
- **O(1)**

```python
class Solution:
    def findComplement(self, num: int) -> int:
        mask = (1 << num.bit_length()) - 1
        return num ^ mask
```
