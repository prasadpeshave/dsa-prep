# Range Sum Query - Immutable

**Problem Link:**  
https://leetcode.com/problems/range-sum-query-immutable/description/

## Problem Statement
Given an integer array `nums`, handle multiple queries of the following type:
- Calculate the sum of the elements of `nums` between indices `left` and `right` inclusive where `left <= right`.

Implement the `NumArray` class:
- `NumArray(int[] nums)` — Initializes the object with the integer array `nums`.
- `int sumRange(int left, int right)` — Returns the sum of elements between indices `left` and `right` inclusive.

## Examples

### Example 1
```
Input:
["NumArray", "sumRange", "sumRange", "sumRange"]
[[[-2, 0, 3, -5, 2, -1]], [0, 2], [2, 5], [0, 5]]

Output: [null, 1, -1, -3]

Explanation:
numArray = NumArray([-2, 0, 3, -5, 2, -1])
numArray.sumRange(0, 2)  # (-2) + 0 + 3        = 1
numArray.sumRange(2, 5)  # 3 + (-5) + 2 + (-1) = -1
numArray.sumRange(0, 5)  # (-2)+0+3+(-5)+2+(-1) = -3
```

## Constraints
- `1 <= nums.length <= 10^4`
- `-10^5 <= nums[i] <= 10^5`
- `0 <= left <= right < nums.length`
- At most `10^4` calls will be made to `sumRange`


## Prefix Sum

### Strategy
- A naive approach recomputes the sum over `[left, right]` on every query in **O(n)** — too slow for `10^4` calls.
- Instead, precompute a **prefix sum array** `prefix` of length `n + 1` where `prefix[i]` stores the sum of the first `i` elements (`nums[0..i-1]`).
  - `prefix[0] = 0` (empty prefix)
  - `prefix[i] = prefix[i-1] + nums[i-1]`
- Any range sum is then a simple subtraction:
  `sumRange(left, right) = prefix[right + 1] - prefix[left]`

### Time Complexity
- **O(n)** to build the prefix array
- **O(1)** per `sumRange` query

### Space Complexity
- **O(n)** for the prefix sum array


````python
from typing import List

class NumArray:

    def __init__(self, nums: List[int]):
        self.prefix = [0] * (len(nums) + 1)
        for i, val in enumerate(nums):
            self.prefix[i + 1] = self.prefix[i] + val

    def sumRange(self, left: int, right: int) -> int:
        return self.prefix[right + 1] - self.prefix[left]


# Your NumArray object will be instantiated and called as such:
# obj = NumArray(nums)
# param_1 = obj.sumRange(left, right)
````

````python
def test_num_array():
    numArray = NumArray([-2, 0, 3, -5, 2, -1])
    assert numArray.sumRange(0, 2) == 1,  f"Expected 1,  got {numArray.sumRange(0, 2)}"
    assert numArray.sumRange(2, 5) == -1, f"Expected -1, got {numArray.sumRange(2, 5)}"
    assert numArray.sumRange(0, 5) == -3, f"Expected -3, got {numArray.sumRange(0, 5)}"

    # Single element queries
    single = NumArray([5])
    assert single.sumRange(0, 0) == 5, f"Expected 5, got {single.sumRange(0, 0)}"

    # All positive
    pos = NumArray([1, 2, 3, 4, 5])
    assert pos.sumRange(0, 4) == 15, f"Expected 15, got {pos.sumRange(0, 4)}"
    assert pos.sumRange(1, 3) == 9,  f"Expected 9,  got {pos.sumRange(1, 3)}"

    # All same
    same = NumArray([3, 3, 3, 3])
    assert same.sumRange(0, 3) == 12, f"Expected 12, got {same.sumRange(0, 3)}"
    assert same.sumRange(1, 2) == 6,  f"Expected 6,  got {same.sumRange(1, 2)}"

    print("All test cases passed!")

test_num_array()
````
