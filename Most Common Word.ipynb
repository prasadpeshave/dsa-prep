# Most Common Word

## Problem Statement

Given a paragraph and a list of banned words, return the most frequent word that is not banned.

- Words are case-insensitive.
- Punctuation should be ignored.
- The result must be returned in lowercase.

It is guaranteed that the answer exists and is unique.

---

## Example

Input:
paragraph = "Bob hit a ball, the hit BALL flew far after it was hit."
banned = ["hit"]

Output:
"ball"

# Approach
---

## HashMap + String Cleaning

### Strategy
1. Convert paragraph to lowercase.
2. Replace punctuation with spaces.
3. Split into words.
4. Use a frequency dictionary (Counter).
5. Ignore banned words while counting.
6. Return the word with highest frequency.

### Time Complexity
- **O(n)** where n is paragraph length

### Space Complexity
- **O(n)**

```python
from typing import List
from collections import Counter
import re

class Solution:
    def mostCommonWord(self, paragraph: str, banned: List[str]) -> str:
        banned_set = set(banned)

        words = re.sub(r'[^a-zA-Z]', ' ', paragraph.lower()).split()

        freq = Counter()

        for word in words:
            if word not in banned_set:
                freq[word] += 1

        return max(freq, key=freq.get)
```

# Test Cases
---

```python
def test_most_common_word():
    sol = Solution()

    paragraph1 = "Bob hit a ball, the hit BALL flew far after it was hit."
    banned1 = ["hit"]
    assert sol.mostCommonWord(paragraph1, banned1) == "ball"

    paragraph2 = "a."
    banned2 = []
    assert sol.mostCommonWord(paragraph2, banned2) == "a"

    print("All test cases passed!")

test_most_common_word()
```
