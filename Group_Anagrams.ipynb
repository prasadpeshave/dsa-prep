# Group Anagrams

**Problem Link:**  
https://leetcode.com/problems/group-anagrams/description/

## Problem Statement
Given an array of strings `strs`, group the anagrams together. You can return the answer in any order.

## Examples

### Example 1
Input: `strs = ["eat","tea","tan","ate","nat","bat"]`  
Output: `[["bat"],["nat","tan"],["ate","eat","tea"]]`  
Explanation:
- `"bat"` has no anagram in the list
- `"nat"` and `"tan"` are anagrams of each other
- `"ate"`, `"eat"`, and `"tea"` are anagrams of each other

### Example 2
Input: `strs = [""]`  
Output: `[[""]]`

### Example 3
Input: `strs = ["a"]`  
Output: `[["a"]]`

## Constraints
- `1 <= strs.length <= 10⁴`
- `0 <= strs[i].length <= 100`
- `strs[i]` consists of lowercase English letters


## Hash Map with Sorted Key

### Strategy
- Two strings are anagrams if and only if their **sorted characters are identical**
- Use a dictionary where the **key** is the sorted version of each string (as a tuple), and the **value** is the list of strings that share that key
- Iterate through `strs`, compute each string's sorted key, and append the string to its corresponding group
- Return all dictionary values as the result

### Why tuple instead of string as key?
- Tuples are hashable and work directly as dictionary keys
- `tuple(sorted(s))` is slightly more idiomatic than `"".join(sorted(s))` but both work equally well

### Time Complexity
- **O(n × k log k)** where `n` is the number of strings and `k` is the maximum string length — sorting each string costs O(k log k)

### Space Complexity
- **O(n × k)** for the hash map storing all strings


```python
from typing import List
from collections import defaultdict

class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        groups = defaultdict(list)

        for s in strs:
            key = tuple(sorted(s))
            groups[key].append(s)

        return list(groups.values())
```

```python
def test_group_anagrams():
    sol = Solution()

    def normalize(groups):
        return sorted([sorted(g) for g in groups])

    # Example 1
    assert normalize(sol.groupAnagrams(["eat","tea","tan","ate","nat","bat"])) == \
           normalize([["bat"],["nat","tan"],["ate","eat","tea"]])

    # Example 2: single empty string
    assert normalize(sol.groupAnagrams([""])) == normalize([[""]])

    # Example 3: single character
    assert normalize(sol.groupAnagrams(["a"])) == normalize([["a"]])

    # All strings are anagrams of each other
    assert normalize(sol.groupAnagrams(["abc","bca","cab"])) == normalize([["abc","bca","cab"]])

    # No two strings are anagrams — each in its own group
    result = sol.groupAnagrams(["abc","def","ghi"])
    assert len(result) == 3
    assert all(len(g) == 1 for g in result)

    # Multiple empty strings group together
    assert normalize(sol.groupAnagrams(["","",""])) == normalize([["" ,"",""]])

    # Mixed lengths — different lengths can never be anagrams
    result = sol.groupAnagrams(["a","ab","ba","b"])
    assert normalize(result) == normalize([["a"],["b"],["ab","ba"]])

    print("All test cases passed!")

test_group_anagrams()
```
